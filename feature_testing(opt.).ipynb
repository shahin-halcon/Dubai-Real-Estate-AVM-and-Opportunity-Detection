#### Model 1 Creation

```python
import pandas as pd
import numpy as np

model_df = flat_sales[
    flat_sales["PROCEDURE_EN"].isin([
        "Sell - Pre registration",
        "Sale"
    ])
].copy()

model_df["INSTANCE_DATE"] = pd.to_datetime(
    model_df["INSTANCE_DATE"]
)

model_df["log_area_sqft"] = np.log1p(
    model_df["area_sqft"]
)

model_df["log_price"] = np.log1p(
    model_df["TRANS_VALUE"]
)
```

```python
def extract_bedrooms(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().upper()

    if value == "STUDIO":
        return 0

    if "B/R" in value:
        return float(
            value.replace("B/R", "").strip()
        )

    return np.nan


model_df["bedrooms"] = (
    model_df["ROOMS_EN"]
    .apply(extract_bedrooms)
)
```

```python
model_df["is_offplan"] = (
    model_df["IS_OFFPLAN_EN"] == "Off-Plan"
).astype(int)

model_df["is_freehold"] = (
    model_df["IS_FREE_HOLD_EN"] == "Free Hold"
).astype(int)
```

```python
train = model_df[
    model_df["INSTANCE_DATE"] < "2026-06-01"
].copy()

validation = model_df[
    (model_df["INSTANCE_DATE"] >= "2026-06-01") &
    (model_df["INSTANCE_DATE"] < "2026-07-01")
].copy()

test = model_df[
    model_df["INSTANCE_DATE"] >= "2026-07-01"
].copy()

print("Train:", train.shape)
print("Validation:", validation.shape)
print("Test:", test.shape)
```

```python
project_counts = (
    train["PROJECT_EN"]
    .value_counts()
)

valid_projects = project_counts[
    project_counts >= 20
].index
```

```python
for data in [train, validation, test]:

    data["project_model"] = (
        data["PROJECT_EN"]
        .where(
            data["PROJECT_EN"].isin(valid_projects),
            "OTHER_PROJECT"
        )
        .fillna("UNKNOWN_PROJECT")
    )
```

```python
numeric_features = [
    "log_area_sqft",
    "bedrooms",
    "is_offplan",
    "is_freehold"
]

categorical_features = [
    "AREA_EN",
    "project_model"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    )
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "num",
        numeric_pipeline,
        numeric_features
    ),
    (
        "cat",
        categorical_pipeline,
        categorical_features
    )
])
```

```python
model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "regressor",
        Ridge(alpha=1.0)
    )
])
```

```python
X_train = train[
    numeric_features + categorical_features
]

y_train = train["log_price"]

X_test = test[
    numeric_features + categorical_features
]

y_test = test["log_price"]
```

```python
model.fit(
    X_train,
    y_train
)
```

```python
pred_log = model.predict(X_test)

pred_price = np.expm1(
    pred_log
)

actual_price = test["TRANS_VALUE"]
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(
    actual_price,
    pred_price
)

rmse = np.sqrt(
    mean_squared_error(
        actual_price,
        pred_price
    )
)

r2 = r2_score(
    actual_price,
    pred_price
)

ape = (
    np.abs(
        actual_price - pred_price
    )
    / actual_price
)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
print(
    "Median Absolute Percentage Error:",
    np.median(ape)
)
print(
    "Within 10%:",
    np.mean(ape <= 0.10)
)
print(
    "Within 20%:",
    np.mean(ape <= 0.20)
)
```

```python
def rolling_features(
    df,
    group_col,
    date_col,
    value_col,
    window="90D"
):
    
    temp = df[
        [group_col, date_col, value_col]
    ].copy()

    temp = temp.sort_values(
        [group_col, date_col]
    )

    result = (
        temp
        .set_index(date_col)
        .groupby(group_col)[value_col]
        .rolling(
            window=window,
            closed="left"
        )
        .agg(["median", "count"])
        .reset_index()
    )

    result = result.rename(
        columns={
            "median": f"{group_col}_90d_median_ppsf",
            "count": f"{group_col}_90d_transaction_count"
        }
    )

    return result
```

```python
# ============================================================
# MODEL 1A — BASELINE RIDGE
# SAME DATA / SAME SPLIT AS MODEL 1B-L
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 1. Baseline features
# ------------------------------------------------------------

baseline_numeric = [
    "log_area_sqft",
    "bedrooms",
    "parking_count",
    "is_offplan",
    "is_freehold"
]

baseline_categorical = [
    "AREA_EN",
    "PROJECT_EN"
]

baseline_features = (
    baseline_numeric +
    baseline_categorical
)


# ------------------------------------------------------------
# 2. X / y
# ------------------------------------------------------------

X_base = hist_df[
    baseline_features
].copy()

y_log_base = hist_df[
    "log_price"
].copy()


# ------------------------------------------------------------
# 3. Same chronological split
# ------------------------------------------------------------

train_end = 55327
val_end = 55327 + 11136

X_train_base = X_base.iloc[
    :train_end
].copy()

X_val_base = X_base.iloc[
    train_end:val_end
].copy()

X_test_base = X_base.iloc[
    val_end:
].copy()

y_train_base = y_log_base.iloc[
    :train_end
].copy()

y_val_base = y_log_base.iloc[
    train_end:val_end
].copy()

y_test_base = y_log_base.iloc[
    val_end:
].copy()


print("Train:", X_train_base.shape)
print("Validation:", X_val_base.shape)
print("Test:", X_test_base.shape)


# ------------------------------------------------------------
# 4. Preprocessing
# ------------------------------------------------------------

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    )
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

baseline_preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_pipeline,
        baseline_numeric
    ),
    (
        "categorical",
        categorical_pipeline,
        baseline_categorical
    )
])


# ------------------------------------------------------------
# 5. Model
# ------------------------------------------------------------

model_1a = Pipeline([
    (
        "preprocessor",
        baseline_preprocessor
    ),
    (
        "model",
        Ridge(alpha=10.0)
    )
])


# ------------------------------------------------------------
# 6. Train
# ------------------------------------------------------------

model_1a.fit(
    X_train_base,
    y_train_base
)


# ------------------------------------------------------------
# 7. Predict
# ------------------------------------------------------------

pred_log_base = model_1a.predict(
    X_test_base
)

pred_base = np.expm1(
    pred_log_base
)

actual_base = np.expm1(
    y_test_base
)


# ------------------------------------------------------------
# 8. Evaluate
# ------------------------------------------------------------

mae_base = mean_absolute_error(
    actual_base,
    pred_base
)

rmse_base = np.sqrt(
    mean_squared_error(
        actual_base,
        pred_base
    )
)

r2_base = r2_score(
    actual_base,
    pred_base
)

ape_base = np.abs(
    (actual_base - pred_base) /
    actual_base
)

mdape_base = np.median(
    ape_base
)

within10_base = np.mean(
    ape_base <= 0.10
)

within20_base = np.mean(
    ape_base <= 0.20
)


# ------------------------------------------------------------
# 9. Results
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MODEL 1A — BASELINE RIDGE")
print("=" * 60)

print(f"MAE:                     AED {mae_base:,.2f}")
print(f"RMSE:                    AED {rmse_base:,.2f}")
print(f"R²:                      {r2_base:.4f}")
print(f"Median Absolute % Error: {mdape_base:.4f}")
print(f"Within ±10%:             {within10_base:.4f}")
print(f"Within ±20%:             {within20_base:.4f}")
```

```python
# ============================================================
# MODEL 2 — XGBOOST AVM
# SAME FEATURES / SAME SPLIT AS MODEL 1B-L
# LOG-TRANSFORMED TARGET
# ============================================================

import xgboost as xgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 1. Features
# ------------------------------------------------------------

numeric_features = [
    "log_area_sqft",
    "bedrooms",
    "parking_count",
    "is_offplan",
    "is_freehold",
    "area_90d_median_ppsf",
    "project_90d_median_ppsf",
    "log_area_90d_count",
    "log_project_90d_count",
    "project_history_available"
]

categorical_features = [
    "AREA_EN",
    "PROJECT_EN"
]

all_features = (
    numeric_features +
    categorical_features
)


# ------------------------------------------------------------
# 2. Prepare data
# ------------------------------------------------------------

xgb_df = hist_df.copy()

X = xgb_df[all_features].copy()
y = xgb_df["log_price"].copy()


# ------------------------------------------------------------
# 3. Encode categorical variables
#
# XGBoost cannot directly consume arbitrary object columns
# in the setup we're using, so use category codes.
#
# IMPORTANT:
# Unknown categories are handled consistently.
# ------------------------------------------------------------

for col in categorical_features:

    X[col] = (
        X[col]
        .astype("category")
        .cat.codes
    )

    # -1 represents missing category
    X[col] = X[col].replace(-1, np.nan)


# ------------------------------------------------------------
# 4. Chronological split
# ------------------------------------------------------------

train_end = 55327
val_end = 55327 + 11136

X_train = X.iloc[:train_end].copy()
X_val   = X.iloc[train_end:val_end].copy()
X_test  = X.iloc[val_end:].copy()

y_train = y.iloc[:train_end].copy()
y_val   = y.iloc[train_end:val_end].copy()
y_test  = y.iloc[val_end:].copy()


print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)


# ------------------------------------------------------------
# 5. XGBoost model
# ------------------------------------------------------------

xgb_model = xgb.XGBRegressor(

    objective="reg:squarederror",

    n_estimators=2000,

    learning_rate=0.03,

    max_depth=6,

    min_child_weight=10,

    subsample=0.8,

    colsample_bytree=0.8,

    reg_alpha=0.1,

    reg_lambda=5.0,

    random_state=42,

    n_jobs=-1,

    early_stopping_rounds=75
)


# ------------------------------------------------------------
# 6. Train
# ------------------------------------------------------------

xgb_model.fit(
    X_train,
    y_train,

    eval_set=[
        (X_train, y_train),
        (X_val, y_val)
    ],

    verbose=False
)


# ------------------------------------------------------------
# 7. Best iteration
# ------------------------------------------------------------

print(
    "\nBest iteration:",
    xgb_model.best_iteration
)

print(
    "Best validation score:",
    xgb_model.best_score
)


# ------------------------------------------------------------
# 8. Predict test set
# ------------------------------------------------------------

pred_log_xgb = xgb_model.predict(
    X_test
)

pred_xgb = np.expm1(
    pred_log_xgb
)

actual_xgb = np.expm1(
    y_test
)


# ------------------------------------------------------------
# 9. Evaluation
# ------------------------------------------------------------

mae_xgb = mean_absolute_error(
    actual_xgb,
    pred_xgb
)

rmse_xgb = np.sqrt(
    mean_squared_error(
        actual_xgb,
        pred_xgb
    )
)

r2_xgb = r2_score(
    actual_xgb,
    pred_xgb
)

ape_xgb = np.abs(
    (actual_xgb - pred_xgb) /
    actual_xgb
)

mdape_xgb = np.median(
    ape_xgb
)

within10_xgb = np.mean(
    ape_xgb <= 0.10
)

within20_xgb = np.mean(
    ape_xgb <= 0.20
)


# ------------------------------------------------------------
# 10. Results
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("MODEL 2 — XGBOOST AVM")
print("=" * 65)

print(
    f"MAE:                     AED {mae_xgb:,.2f}"
)

print(
    f"RMSE:                    AED {rmse_xgb:,.2f}"
)

print(
    f"R²:                      {r2_xgb:.4f}"
)

print(
    f"Median Absolute % Error: {mdape_xgb:.4f}"
)

print(
    f"Within ±10%:             {within10_xgb:.4f}"
)

print(
    f"Within ±20%:             {within20_xgb:.4f}"
)
```

```python
# ============================================================
# STEP 10C — 30-DAY FORWARD PROJECT COMPARABLE PPSF
# ============================================================

# Target transactions: every transaction in bt_clean
targets = bt_clean[
    [
        "TRANSACTION_NUMBER",
        "INSTANCE_DATE",
        "PROJECT_EN",
        "ROOMS_EN",
        "IS_OFFPLAN_EN",
        "price_per_sqft"
    ]
].copy()

targets = targets.rename(
    columns={
        "INSTANCE_DATE": "target_date",
        "price_per_sqft": "current_ppsf"
    }
)

# Future transaction universe
future = bt_clean[
    [
        "INSTANCE_DATE",
        "PROJECT_EN",
        "ROOMS_EN",
        "IS_OFFPLAN_EN",
        "price_per_sqft"
    ]
].copy()

future = future.rename(
    columns={
        "INSTANCE_DATE": "future_date",
        "price_per_sqft": "future_ppsf"
    }
)

# ------------------------------------------------------------
# Merge target transactions with transactions from the same
# project.
#
# This is deliberately done in a simple, transparent way.
# We'll restrict the result to a 30-day forward window.
# ------------------------------------------------------------

merged = targets.merge(
    future,
    on=["PROJECT_EN", "ROOMS_EN", "IS_OFFPLAN_EN"],
    how="left"
)

# Only future transactions strictly after the target
merged = merged[
    merged["future_date"] > merged["target_date"]
].copy()

# Calculate days forward
merged["days_forward"] = (
    merged["future_date"]
    - merged["target_date"]
).dt.total_seconds() / (60 * 60 * 24)

# Keep only the next 30 days
merged_30d = merged[
    merged["days_forward"] <= 30
].copy()

# ------------------------------------------------------------
# Aggregate future comparable PPSF
# ------------------------------------------------------------

future_30d = (
    merged_30d
    .groupby(
        [
            "TRANSACTION_NUMBER",
            "target_date"
        ],
        as_index=False
    )
    .agg(
        future_30d_ppsf=("future_ppsf", "median"),
        future_30d_comps=("future_ppsf", "size")
    )
)

print("Target transactions:", len(targets))
print("Transactions with 30-day future comps:", len(future_30d))

print("\nFuture comparable count:")
print(
    future_30d["future_30d_comps"].describe()
)

print("\nFuture PPSF:")
print(
    future_30d["future_30d_ppsf"].describe()
)
```

```python
# ============================================================
# STEP 10D — 30-DAY FORWARD RETURN
# ============================================================

# Merge the forward comparable result back to the target data
backtest_30d = targets.merge(
    future_30d,
    on=["TRANSACTION_NUMBER", "target_date"],
    how="left"
)

# Require at least 3 future comparable transactions
backtest_30d["has_30d_signal"] = (
    backtest_30d["future_30d_comps"] >= 3
)

# Forward PPSF return
backtest_30d["forward_30d_return_pct"] = (
    (
        backtest_30d["future_30d_ppsf"]
        / backtest_30d["current_ppsf"]
    ) - 1
) * 100

# Keep only observations with >= 3 future comps
valid_30d = backtest_30d[
    backtest_30d["has_30d_signal"]
].copy()

print(
    "Transactions with >=3 future comparables:",
    len(valid_30d)
)

print("\nForward 30-day return:")
print(
    valid_30d["forward_30d_return_pct"].describe(
        percentiles=[
            .10, .25, .50, .75, .90
        ]
    )
)
```

```python
# ============================================================
# STEP 14 — STATISTICAL VALIDATION + PROJECT CONCENTRATION
# ============================================================

import numpy as np
import pandas as pd

bt14 = backtest_30d.copy()

# ------------------------------------------------------------
# 14A — Prepare data
# ------------------------------------------------------------

bt14["is_high_opportunity"] = (
    bt14["opportunity"]
    .astype(str)
    .str.upper()
    .eq("HIGH OPPORTUNITY")
)

bt14 = bt14[
    bt14["forward_30d_return_pct"].notna()
].copy()

high14 = bt14[
    bt14["is_high_opportunity"]
].copy()

non_high14 = bt14[
    ~bt14["is_high_opportunity"]
].copy()

print("============================================================")
print("STEP 14 — STATISTICAL VALIDATION")
print("============================================================")

print(f"Total observations: {len(bt14):,}")
print(f"HIGH OPPORTUNITY:   {len(high14):,}")
print(f"NON-HIGH:           {len(non_high14):,}")


# ------------------------------------------------------------
# 14B — Bootstrap confidence intervals
# ------------------------------------------------------------

rng = np.random.default_rng(42)

def bootstrap_ci(values, statistic, n_boot=10000):

    values = np.asarray(values, dtype=float)

    boot_stats = np.empty(n_boot)

    for i in range(n_boot):
        sample = rng.choice(
            values,
            size=len(values),
            replace=True
        )
        boot_stats[i] = statistic(sample)

    lower = np.percentile(boot_stats, 2.5)
    upper = np.percentile(boot_stats, 97.5)

    return lower, upper


high_returns = high14[
    "forward_30d_return_pct"
].to_numpy()

non_high_returns = non_high14[
    "forward_30d_return_pct"
].to_numpy()


high_mean_ci = bootstrap_ci(
    high_returns,
    np.mean
)

high_median_ci = bootstrap_ci(
    high_returns,
    np.median
)

non_high_mean_ci = bootstrap_ci(
    non_high_returns,
    np.mean
)

non_high_median_ci = bootstrap_ci(
    non_high_returns,
    np.median
)


print("\n============================================================")
print("14B — BOOTSTRAP 95% CONFIDENCE INTERVALS")
print("============================================================")

print(
    f"HIGH mean return: "
    f"{high_returns.mean():.2f}% "
    f"[{high_mean_ci[0]:.2f}%, {high_mean_ci[1]:.2f}%]"
)

print(
    f"HIGH median return: "
    f"{np.median(high_returns):.2f}% "
    f"[{high_median_ci[0]:.2f}%, {high_median_ci[1]:.2f}%]"
)

print(
    f"NON-HIGH mean return: "
    f"{non_high_returns.mean():.2f}% "
    f"[{non_high_mean_ci[0]:.2f}%, {non_high_mean_ci[1]:.2f}%]"
)

print(
    f"NON-HIGH median return: "
    f"{np.median(non_high_returns):.2f}% "
    f"[{non_high_median_ci[0]:.2f}%, {non_high_median_ci[1]:.2f}%]"
)


# ------------------------------------------------------------
# 14C — Bootstrap distribution of mean lift
# ------------------------------------------------------------

n_boot = 10000

lift_boot = np.empty(n_boot)

for i in range(n_boot):

    high_sample = rng.choice(
        high_returns,
        size=len(high_returns),
        replace=True
    )

    baseline_sample = rng.choice(
        non_high_returns,
        size=len(non_high_returns),
        replace=True
    )

    lift_boot[i] = (
        np.mean(high_sample)
        - np.mean(baseline_sample)
    )

lift_ci = np.percentile(
    lift_boot,
    [2.5, 50, 97.5]
)

print("\n============================================================")
print("14C — BOOTSTRAP MEAN-RETURN LIFT")
print("============================================================")

print(
    f"Observed mean lift: "
    f"{high_returns.mean() - non_high_returns.mean():.2f} pp"
)

print(
    f"Bootstrap median lift: "
    f"{lift_ci[1]:.2f} pp"
)

print(
    f"95% CI: "
    f"[{lift_ci[0]:.2f}, {lift_ci[2]:.2f}] pp"
)


# ------------------------------------------------------------
# 14D — Leave RAW DISTRICT out
# ------------------------------------------------------------

dominant_project = "RAW DISTRICT BY IMTIAZ R"

bt_no_raw = bt14[
    bt14["PROJECT_EN"].astype(str).str.upper()
    != dominant_project.upper()
].copy()

high_no_raw = bt_no_raw[
    bt_no_raw["is_high_opportunity"]
]

non_high_no_raw = bt_no_raw[
    ~bt_no_raw["is_high_opportunity"]
]

print("\n============================================================")
print("14D — LEAVE-ONE-PROJECT-OUT TEST")
print("============================================================")

print(
    f"Removed project: {dominant_project}"
)

print(
    f"Remaining HIGH OPPORTUNITY: "
    f"{len(high_no_raw)}"
)

if len(high_no_raw):

    r = high_no_raw[
        "forward_30d_return_pct"
    ]

    print(
        f"Median return: "
        f"{r.median():.2f}%"
    )

    print(
        f"Mean return: "
        f"{r.mean():.2f}%"
    )

    print(
        f"Positive return: "
        f"{(r > 0).mean() * 100:.2f}%"
    )

    print(
        f">10% return: "
        f"{(r > 10).mean() * 100:.2f}%"
    )

else:

    print(
        "No HIGH OPPORTUNITY observations remain."
    )


# ------------------------------------------------------------
# 14E — Project-level performance
# ------------------------------------------------------------

print("\n============================================================")
print("14E — PROJECT-LEVEL SIGNAL PERFORMANCE")
print("============================================================")

project_stats = (
    high14
    .groupby("PROJECT_EN")
    ["forward_30d_return_pct"]
    .agg(
        transactions="count",
        median_return="median",
        mean_return="mean",
        positive_rate=lambda x: (x > 0).mean() * 100,
        above_10=lambda x: (x > 10).mean() * 100,
    )
    .sort_values(
        ["transactions", "median_return"],
        ascending=[False, False]
    )
)

print(
    project_stats.to_string()
)


# ------------------------------------------------------------
# 14F — Final interpretation numbers
# ------------------------------------------------------------

print("\n============================================================")
print("STEP 14 — COMPLETE")
print("============================================================")

print(
    f"HIGH median: "
    f"{np.median(high_returns):.2f}%"
)

print(
    f"HIGH median 95% CI: "
    f"[{high_median_ci[0]:.2f}%, "
    f"{high_median_ci[1]:.2f}%]"
)

print(
    f"HIGH mean: "
    f"{high_returns.mean():.2f}%"
)

print(
    f"HIGH mean 95% CI: "
    f"[{high_mean_ci[0]:.2f}%, "
    f"{high_mean_ci[1]:.2f}%]"
)

print(
    f"Mean lift 95% CI: "
    f"[{lift_ci[0]:.2f}, {lift_ci[2]:.2f}] pp"
)

print(
    f"HIGH observations excluding RAW DISTRICT: "
    f"{len(high_no_raw)}"
)
```

```python
# ============================================================
# STEP 15 — SIGNAL QUALITY / THRESHOLD ANALYSIS
# ============================================================

import numpy as np
import pandas as pd

bt15 = backtest_30d.copy()

bt15["is_high_opportunity"] = (
    bt15["opportunity"]
    .astype(str)
    .str.upper()
    .eq("HIGH OPPORTUNITY")
)

bt15 = bt15[
    bt15["forward_30d_return_pct"].notna()
].copy()

print("============================================================")
print("STEP 15 — SIGNAL QUALITY / THRESHOLD ANALYSIS")
print("============================================================")

print(f"Backtest rows: {len(bt15):,}")
print(
    f"HIGH OPPORTUNITY rows: "
    f"{bt15['is_high_opportunity'].sum():,}"
)


# ------------------------------------------------------------
# 15A — HIGH OPPORTUNITY score distribution
# ------------------------------------------------------------

high15 = bt15[
    bt15["is_high_opportunity"]
].copy()

print("\n============================================================")
print("15A — HIGH OPPORTUNITY SCORE DISTRIBUTION")
print("============================================================")

print(
    high15[
        [
            "opportunity_score",
            "confidence_score",
            "risk_adjusted_gap_pct"
        ]
    ].describe().to_string()
)


# ------------------------------------------------------------
# 15B — Threshold analysis
#
# IMPORTANT:
# These thresholds use ONLY variables available at the
# transaction date.
# ------------------------------------------------------------

thresholds = [
    ("score >= 0.55", bt15["opportunity_score"] >= 0.55),
    ("score >= 0.60", bt15["opportunity_score"] >= 0.60),
    ("score >= 0.65", bt15["opportunity_score"] >= 0.65),
    ("score >= 0.70", bt15["opportunity_score"] >= 0.70),
    ("score >= 0.75", bt15["opportunity_score"] >= 0.75),
    ("confidence >= 60", bt15["confidence_score"] >= 60),
    ("confidence >= 70", bt15["confidence_score"] >= 70),
    ("confidence >= 80", bt15["confidence_score"] >= 80),
    ("gap >= 20%", bt15["risk_adjusted_gap_pct"] >= 20),
    ("gap >= 25%", bt15["risk_adjusted_gap_pct"] >= 25),
    ("gap >= 30%", bt15["risk_adjusted_gap_pct"] >= 30),
]


rows = []

for label, mask in thresholds:

    subset = bt15[mask]

    if len(subset) == 0:
        continue

    r = subset["forward_30d_return_pct"]

    rows.append({
        "rule": label,
        "transactions": len(subset),
        "median_return_pct": r.median(),
        "mean_return_pct": r.mean(),
        "pct_positive": (r > 0).mean() * 100,
        "pct_above_10pct": (r > 10).mean() * 100,
    })


threshold_results = pd.DataFrame(rows)

print("\n============================================================")
print("15B — INDIVIDUAL SIGNAL THRESHOLDS")
print("============================================================")

print(
    threshold_results.to_string(index=False)
)


# ------------------------------------------------------------
# 15C — HIGH OPPORTUNITY + confidence
# ------------------------------------------------------------

print("\n============================================================")
print("15C — HIGH OPPORTUNITY + CONFIDENCE")
print("============================================================")

confidence_rows = []

for threshold in [50, 60, 65, 70, 75, 80]:

    subset = high15[
        high15["confidence_score"] >= threshold
    ]

    if len(subset) == 0:
        continue

    r = subset["forward_30d_return_pct"]

    confidence_rows.append({
        "confidence_min": threshold,
        "transactions": len(subset),
        "median_return_pct": r.median(),
        "mean_return_pct": r.mean(),
        "pct_positive": (r > 0).mean() * 100,
        "pct_above_10pct": (r > 10).mean() * 100,
    })

confidence_results = pd.DataFrame(
    confidence_rows
)

print(
    confidence_results.to_string(index=False)
)


# ------------------------------------------------------------
# 15D — HIGH OPPORTUNITY + risk-adjusted gap
# ------------------------------------------------------------

print("\n============================================================")
print("15D — HIGH OPPORTUNITY + GAP")
print("============================================================")

gap_rows = []

for threshold in [10, 15, 20, 25, 30, 35, 40]:

    subset = high15[
        high15["risk_adjusted_gap_pct"] >= threshold
    ]

    if len(subset) == 0:
        continue

    r = subset["forward_30d_return_pct"]

    gap_rows.append({
        "gap_min_pct": threshold,
        "transactions": len(subset),
        "median_return_pct": r.median(),
        "mean_return_pct": r.mean(),
        "pct_positive": (r > 0).mean() * 100,
        "pct_above_10pct": (r > 10).mean() * 100,
    })

gap_results = pd.DataFrame(gap_rows)

print(
    gap_results.to_string(index=False)
)


# ------------------------------------------------------------
# 15E — Combined conservative rules
# ------------------------------------------------------------

print("\n============================================================")
print("15E — COMBINED CONSERVATIVE RULES")
print("============================================================")

combined_rules = [
    (
        "HIGH + confidence >= 70",
        high15["confidence_score"] >= 70
    ),
    (
        "HIGH + confidence >= 80",
        high15["confidence_score"] >= 80
    ),
    (
        "HIGH + gap >= 20%",
        high15["risk_adjusted_gap_pct"] >= 20
    ),
    (
        "HIGH + gap >= 25%",
        high15["risk_adjusted_gap_pct"] >= 25
    ),
    (
        "HIGH + gap >= 30%",
        high15["risk_adjusted_gap_pct"] >= 30
    ),
    (
        "HIGH + score >= 0.65",
        high15["opportunity_score"] >= 0.65
    ),
    (
        "HIGH + score >= 0.70",
        high15["opportunity_score"] >= 0.70
    ),
]


combined_rows = []

for label, condition in combined_rules:

    subset = high15[condition]

    if len(subset) == 0:
        continue

    r = subset["forward_30d_return_pct"]

    combined_rows.append({
        "rule": label,
        "transactions": len(subset),
        "median_return_pct": r.median(),
        "mean_return_pct": r.mean(),
        "pct_positive": (r > 0).mean() * 100,
        "pct_above_10pct": (r > 10).mean() * 100,
    })

combined_results = pd.DataFrame(
    combined_rows
)

print(
    combined_results.to_string(index=False)
)


# ------------------------------------------------------------
# 15F — Project robustness for the strongest practical rule
# ------------------------------------------------------------

print("\n============================================================")
print("15F — PROJECT ROBUSTNESS")
print("============================================================")

if len(combined_results):

    best_rule = combined_results.sort_values(
        [
            "median_return_pct",
            "transactions"
        ],
        ascending=[False, False]
    ).iloc[0]

    print("Highest median-return rule:")
    print(best_rule.to_string())

print("\n============================================================")
print("STEP 15 — COMPLETE")
print("============================================================")
```

```python
# ============================================================
# STEP 16 — TEMPORAL HOLDOUT VALIDATION
# ============================================================

import pandas as pd
import numpy as np

bt16 = backtest_30d.copy()

bt16["is_high_opportunity"] = (
    bt16["opportunity"]
    .astype(str)
    .str.upper()
    .eq("HIGH OPPORTUNITY")
)

bt16 = bt16[
    bt16["forward_30d_return_pct"].notna()
].copy()

bt16["target_date"] = pd.to_datetime(bt16["target_date"])

print("=" * 60)
print("STEP 16 — TEMPORAL HOLDOUT VALIDATION")
print("=" * 60)

print("Date range:")
print(bt16["target_date"].min())
print(bt16["target_date"].max())


# ------------------------------------------------------------
# 16A — Define temporal split
#
# Use the first 70% for rule selection and last 30%
# strictly as holdout.
# ------------------------------------------------------------

split_date = bt16["target_date"].quantile(0.70)

train_bt = bt16[
    bt16["target_date"] < split_date
].copy()

holdout_bt = bt16[
    bt16["target_date"] >= split_date
].copy()

print("\nSplit date:", split_date)

print("Rule-selection period:", len(train_bt))
print("Holdout period:", len(holdout_bt))

print(
    "\nRule-selection dates:",
    train_bt["target_date"].min(),
    "->",
    train_bt["target_date"].max()
)

print(
    "Holdout dates:",
    holdout_bt["target_date"].min(),
    "->",
    holdout_bt["target_date"].max()
)


# ------------------------------------------------------------
# 16B — Candidate rules
# ------------------------------------------------------------

def candidate_rules(df):

    return {
        "HIGH OPPORTUNITY":
            df["is_high_opportunity"],

        "HIGH + confidence >= 70":
            (
                df["is_high_opportunity"]
                & (df["confidence_score"] >= 70)
            ),

        "HIGH + confidence >= 80":
            (
                df["is_high_opportunity"]
                & (df["confidence_score"] >= 80)
            ),

        "HIGH + score >= 0.65":
            (
                df["is_high_opportunity"]
                & (df["opportunity_score"] >= 0.65)
            ),
    }


# ------------------------------------------------------------
# 16C — Evaluation helper
# ------------------------------------------------------------

def evaluate_rule(df, mask):

    subset = df[mask].copy()

    if len(subset) == 0:
        return {
            "transactions": 0,
            "median_return_pct": np.nan,
            "mean_return_pct": np.nan,
            "pct_positive": np.nan,
            "pct_above_5pct": np.nan,
            "pct_above_10pct": np.nan,
        }

    r = subset["forward_30d_return_pct"]

    return {
        "transactions": len(subset),
        "median_return_pct": r.median(),
        "mean_return_pct": r.mean(),
        "pct_positive": (r > 0).mean() * 100,
        "pct_above_5pct": (r > 5).mean() * 100,
        "pct_above_10pct": (r > 10).mean() * 100,
    }


# ------------------------------------------------------------
# 16D — Evaluate selection period
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("16D — RULE SELECTION PERIOD")
print("=" * 60)

selection_rows = []

for name, mask in candidate_rules(train_bt).items():

    result = evaluate_rule(train_bt, mask)

    result["rule"] = name

    selection_rows.append(result)

selection_results = pd.DataFrame(selection_rows)

selection_results = selection_results[
    [
        "rule",
        "transactions",
        "median_return_pct",
        "mean_return_pct",
        "pct_positive",
        "pct_above_5pct",
        "pct_above_10pct",
    ]
]

print(selection_results.to_string(index=False))


# ------------------------------------------------------------
# 16E — Evaluate holdout period
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("16E — TRUE HOLDOUT PERIOD")
print("=" * 60)

holdout_rows = []

for name, mask in candidate_rules(holdout_bt).items():

    result = evaluate_rule(holdout_bt, mask)

    result["rule"] = name

    holdout_rows.append(result)

holdout_results = pd.DataFrame(holdout_rows)

holdout_results = holdout_results[
    [
        "rule",
        "transactions",
        "median_return_pct",
        "mean_return_pct",
        "pct_positive",
        "pct_above_5pct",
        "pct_above_10pct",
    ]
]

print(holdout_results.to_string(index=False))


# ------------------------------------------------------------
# 16F — Compare against holdout baseline
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("16F — HOLDOUT LIFT")
print("=" * 60)

baseline_mask = ~holdout_bt["is_high_opportunity"]

baseline = evaluate_rule(
    holdout_bt,
    baseline_mask
)

print("\nHoldout NON-HIGH baseline:")
print(pd.Series(baseline).to_string())


lift_rows = []

for _, row in holdout_results.iterrows():

    lift_rows.append({
        "rule": row["rule"],
        "transactions": row["transactions"],
        "median_lift_pp":
            row["median_return_pct"]
            - baseline["median_return_pct"],
        "mean_lift_pp":
            row["mean_return_pct"]
            - baseline["mean_return_pct"],
        "positive_rate_lift_pp":
            row["pct_positive"]
            - baseline["pct_positive"],
        "above_10_lift_pp":
            row["pct_above_10pct"]
            - baseline["pct_above_10pct"],
    })

lift_results = pd.DataFrame(lift_rows)

print(
    lift_results.to_string(index=False)
)


# ------------------------------------------------------------
# 16G — Stability check
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("16G — HIGH OPPORTUNITY HOLDOUT PROJECT BREAKDOWN")
print("=" * 60)

high_holdout = holdout_bt[
    holdout_bt["is_high_opportunity"]
].copy()

if len(high_holdout):

    project_holdout = (
        high_holdout
        .groupby("PROJECT_EN")["forward_30d_return_pct"]
        .agg(
            transactions="count",
            median_return="median",
            mean_return="mean",
            positive_rate=lambda x: (x > 0).mean() * 100
        )
        .sort_values(
            "transactions",
            ascending=False
        )
    )

    print(
        project_holdout.to_string()
    )

else:

    print("No HIGH OPPORTUNITY observations in holdout.")


print("\n" + "=" * 60)
print("STEP 16 — COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# STEP 17 — FINAL SIGNAL SPECIFICATION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 60)
print("STEP 17 — FINAL SIGNAL SPECIFICATION")
print("=" * 60)

# ------------------------------------------------------------
# 17A — Work from the validated backtest
# ------------------------------------------------------------

bt17 = backtest_30d.copy()

bt17["target_date"] = pd.to_datetime(bt17["target_date"])

bt17 = bt17[
    bt17["forward_30d_return_pct"].notna()
].copy()

# Normalize HIGH OPPORTUNITY label
bt17["is_high_opportunity"] = (
    bt17["opportunity"]
    .astype(str)
    .str.upper()
    .eq("HIGH OPPORTUNITY")
)

print("\nBacktest observations:", len(bt17))

# ------------------------------------------------------------
# 17B — Freeze the two production signal tiers
#
# Tier 1 = broad opportunity signal
# Tier 2 = conservative/high-confidence signal
# ------------------------------------------------------------

bt17["signal_broad"] = (
    bt17["is_high_opportunity"]
)

bt17["signal_conservative"] = (
    bt17["is_high_opportunity"]
    & (bt17["confidence_score"] >= 80)
)

# ------------------------------------------------------------
# 17C — Summary function
# ------------------------------------------------------------

def signal_summary(df, mask):

    x = df[mask].copy()

    if len(x) == 0:
        return {
            "transactions": 0,
            "median_return_pct": np.nan,
            "mean_return_pct": np.nan,
            "pct_positive": np.nan,
            "pct_above_5pct": np.nan,
            "pct_above_10pct": np.nan,
            "median_opportunity_score": np.nan,
            "median_confidence": np.nan,
            "median_gap_pct": np.nan,
        }

    r = x["forward_30d_return_pct"]

    return {
        "transactions": len(x),
        "median_return_pct": r.median(),
        "mean_return_pct": r.mean(),
        "pct_positive": (r > 0).mean() * 100,
        "pct_above_5pct": (r > 5).mean() * 100,
        "pct_above_10pct": (r > 10).mean() * 100,
        "median_opportunity_score":
            x["opportunity_score"].median(),
        "median_confidence":
            x["confidence_score"].median(),
        "median_gap_pct":
            x["risk_adjusted_gap_pct"].median(),
    }


# ------------------------------------------------------------
# 17D — Evaluate frozen signals
# ------------------------------------------------------------

summary = []

broad = signal_summary(
    bt17,
    bt17["signal_broad"]
)

broad["signal"] = "BROAD — HIGH OPPORTUNITY"

conservative = signal_summary(
    bt17,
    bt17["signal_conservative"]
)

conservative["signal"] = (
    "CONSERVATIVE — HIGH + CONFIDENCE >= 80"
)

summary.extend([
    broad,
    conservative
])

signal_results = pd.DataFrame(summary)

signal_results = signal_results[
    [
        "signal",
        "transactions",
        "median_return_pct",
        "mean_return_pct",
        "pct_positive",
        "pct_above_5pct",
        "pct_above_10pct",
        "median_opportunity_score",
        "median_confidence",
        "median_gap_pct",
    ]
]

print("\n" + "=" * 60)
print("17D — FROZEN SIGNAL PERFORMANCE")
print("=" * 60)

print(
    signal_results.to_string(index=False)
)


# ------------------------------------------------------------
# 17E — Compare conservative signal against broad signal
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("17E — CONSERVATIVE VS BROAD")
print("=" * 60)

print(
    "Additional filtering:",
    broad["transactions"] -
    conservative["transactions"]
)

print(
    "Median-return improvement:",
    round(
        conservative["median_return_pct"]
        - broad["median_return_pct"],
        2
    ),
    "pp"
)

print(
    "Mean-return improvement:",
    round(
        conservative["mean_return_pct"]
        - broad["mean_return_pct"],
        2
    ),
    "pp"
)

print(
    "Positive-rate improvement:",
    round(
        conservative["pct_positive"]
        - broad["pct_positive"],
        2
    ),
    "pp"
)

print(
    ">10% improvement:",
    round(
        conservative["pct_above_10pct"]
        - broad["pct_above_10pct"],
        2
    ),
    "pp"
)


# ------------------------------------------------------------
# 17F — Signal assignment
#
# This is the final hierarchy:
#
# CONSERVATIVE:
# HIGH OPPORTUNITY + confidence >= 80
#
# BROAD:
# HIGH OPPORTUNITY but confidence < 80
#
# EVERYTHING ELSE:
# NO SIGNAL
# ------------------------------------------------------------

bt17["final_signal"] = "NO SIGNAL"

bt17.loc[
    bt17["signal_broad"],
    "final_signal"
] = "BROAD OPPORTUNITY"

bt17.loc[
    bt17["signal_conservative"],
    "final_signal"
] = "CONSERVATIVE OPPORTUNITY"


# ------------------------------------------------------------
# 17G — Final signal counts
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("17G — FINAL SIGNAL COUNTS")
print("=" * 60)

print(
    bt17["final_signal"]
    .value_counts()
)


# ------------------------------------------------------------
# 17H — Conservative signal project concentration
# ------------------------------------------------------------

conservative_df = bt17[
    bt17["final_signal"]
    == "CONSERVATIVE OPPORTUNITY"
].copy()

print("\n" + "=" * 60)
print("17H — CONSERVATIVE PROJECT CONCENTRATION")
print("=" * 60)

if len(conservative_df):

    project_summary = (
        conservative_df
        .groupby("PROJECT_EN")["forward_30d_return_pct"]
        .agg(
            transactions="count",
            median_return="median",
            mean_return="mean",
            positive_rate=lambda x:
                (x > 0).mean() * 100
        )
        .sort_values(
            ["transactions", "median_return"],
            ascending=[False, False]
        )
    )

    print(
        project_summary.to_string()
    )

else:
    print("No conservative opportunities.")


# ------------------------------------------------------------
# 17I — Final specification
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("17I — FINAL PRODUCTION SPECIFICATION")
print("=" * 60)

print("""
SIGNAL 1 — CONSERVATIVE OPPORTUNITY
-----------------------------------
opportunity == HIGH OPPORTUNITY
AND
confidence_score >= 80

Interpretation:
Highest-confidence subset of the validated opportunity signal.


SIGNAL 2 — BROAD OPPORTUNITY
----------------------------
opportunity == HIGH OPPORTUNITY
AND
confidence_score < 80

Interpretation:
Potential opportunity requiring additional review.


NO SIGNAL
---------
All other transactions.

Interpretation:
Do not treat as an identified arbitrage opportunity.
""")


# ------------------------------------------------------------
# 17J — Freeze the final signal dataframe
# ------------------------------------------------------------

final_signal_columns = [
    "TRANSACTION_NUMBER",
    "target_date",
    "PROJECT_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "current_ppsf",
    "future_30d_ppsf",
    "future_30d_comps",
    "forward_30d_return_pct",
    "opportunity_score",
    "confidence_score",
    "risk_adjusted_gap_pct",
    "final_signal",
]

final_signal_df = bt17[
    final_signal_columns
].copy()

print("\nFinal signal dataframe:")
print(final_signal_df.shape)

print("\n" + "=" * 60)
print("STEP 17 — COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# STEP 18 — PORTFOLIO / ECONOMIC SIMULATION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 60)
print("STEP 18 — PORTFOLIO / ECONOMIC SIMULATION")
print("=" * 60)

# ------------------------------------------------------------
# 18A — Use frozen signal dataframe
# ------------------------------------------------------------

portfolio = final_signal_df.copy()

portfolio["target_date"] = pd.to_datetime(
    portfolio["target_date"]
)

# Only transactions with realized forward returns
portfolio = portfolio[
    portfolio["forward_30d_return_pct"].notna()
].copy()

portfolio["forward_return"] = (
    portfolio["forward_30d_return_pct"] / 100.0
)

portfolio["entry_value"] = (
    portfolio["current_ppsf"]
)

print("\nTotal realized observations:", len(portfolio))


# ------------------------------------------------------------
# 18B — Conservative opportunity portfolio
# ------------------------------------------------------------

conservative = portfolio[
    portfolio["final_signal"]
    == "CONSERVATIVE OPPORTUNITY"
].copy()

broad = portfolio[
    portfolio["final_signal"]
    == "BROAD OPPORTUNITY"
].copy()

print(
    "Conservative opportunities:",
    len(conservative)
)

print(
    "Broad opportunities:",
    len(broad)
)


# ------------------------------------------------------------
# 18C — Equal-weight portfolio return
#
# Every opportunity receives equal capital.
# This avoids a single expensive transaction dominating
# the result.
# ------------------------------------------------------------

if len(conservative):

    equal_weight_return = (
        conservative["forward_return"]
        .mean()
    )

    median_trade_return = (
        conservative["forward_return"]
        .median()
    )

    positive_rate = (
        conservative["forward_return"] > 0
    ).mean()

    print("\n" + "=" * 60)
    print("18C — EQUAL-WEIGHT CONSERVATIVE PORTFOLIO")
    print("=" * 60)

    print(
        "Transactions:",
        len(conservative)
    )

    print(
        "Mean portfolio return:",
        round(equal_weight_return * 100, 2),
        "%"
    )

    print(
        "Median trade return:",
        round(median_trade_return * 100, 2),
        "%"
    )

    print(
        "Positive trades:",
        round(positive_rate * 100, 2),
        "%"
    )


# ------------------------------------------------------------
# 18D — Concentration analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18D — PROJECT CONCENTRATION")
print("=" * 60)

project_counts = (
    conservative["PROJECT_EN"]
    .value_counts()
)

project_share = (
    project_counts
    / len(conservative)
    * 100
)

concentration = pd.DataFrame({
    "transactions": project_counts,
    "share_pct": project_share
})

print(
    concentration.to_string()
)

top_project_share = (
    concentration["share_pct"].iloc[0]
    if len(concentration)
    else np.nan
)

print(
    "\nLargest project concentration:",
    round(top_project_share, 2),
    "%"
)


# ------------------------------------------------------------
# 18E — Project concentration capped at 20%
#
# Maximum 20% of the opportunity portfolio can come from
# one project.
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18E — 20% PROJECT CONCENTRATION CAP")
print("=" * 60)

if len(conservative):

    max_per_project = max(
        1,
        int(np.floor(len(conservative) * 0.20))
    )

    print(
        "Maximum observations per project:",
        max_per_project
    )

    capped = (
        conservative
        .sort_values(
            [
                "PROJECT_EN",
                "target_date"
            ]
        )
        .groupby("PROJECT_EN")
        .head(max_per_project)
        .copy()
    )

    print(
        "Original opportunities:",
        len(conservative)
    )

    print(
        "After concentration cap:",
        len(capped)
    )

    print(
        "Mean return:",
        round(
            capped["forward_return"].mean() * 100,
            2
        ),
        "%"
    )

    print(
        "Median return:",
        round(
            capped["forward_return"].median() * 100,
            2
        ),
        "%"
    )

    print(
        "Positive rate:",
        round(
            (capped["forward_return"] > 0).mean() * 100,
            2
        ),
        "%"
    )

    print(
        ">10% return:",
        round(
            (capped["forward_return"] > 0.10).mean() * 100,
            2
        ),
        "%"
    )

else:

    capped = pd.DataFrame()


# ------------------------------------------------------------
# 18F — Capital scenarios
#
# Illustrative equal allocation:
# AED 1M total capital
# AED 5M total capital
# AED 10M total capital
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18F — CAPITAL SCENARIOS")
print("=" * 60)

capital_levels = [
    1_000_000,
    5_000_000,
    10_000_000,
]

if len(conservative):

    for capital in capital_levels:

        allocation = (
            capital
            / len(conservative)
        )

        simulated_profit = (
            conservative["forward_return"]
            * allocation
        ).sum()

        portfolio_return = (
            simulated_profit
            / capital
        )

        print(
            f"\nCapital: AED {capital:,.0f}"
        )

        print(
            "Equal allocation per opportunity:",
            f"AED {allocation:,.2f}"
        )

        print(
            "Simulated gross profit:",
            f"AED {simulated_profit:,.2f}"
        )

        print(
            "Portfolio return:",
            f"{portfolio_return * 100:.2f}%"
        )


# ------------------------------------------------------------
# 18G — Worst trades
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18G — DOWNSIDE / WORST CONSERVATIVE TRADES")
print("=" * 60)

if len(conservative):

    worst = (
        conservative[
            [
                "TRANSACTION_NUMBER",
                "target_date",
                "PROJECT_EN",
                "ROOMS_EN",
                "current_ppsf",
                "forward_30d_return_pct",
                "opportunity_score",
                "confidence_score",
            ]
        ]
        .sort_values(
            "forward_30d_return_pct"
        )
        .head(10)
    )

    print(
        worst.to_string(index=False)
    )


# ------------------------------------------------------------
# 18H — Best trades
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18H — BEST CONSERVATIVE TRADES")
print("=" * 60)

if len(conservative):

    best = (
        conservative[
            [
                "TRANSACTION_NUMBER",
                "target_date",
                "PROJECT_EN",
                "ROOMS_EN",
                "current_ppsf",
                "forward_30d_return_pct",
                "opportunity_score",
                "confidence_score",
            ]
        ]
        .sort_values(
            "forward_30d_return_pct",
            ascending=False
        )
        .head(10)
    )

    print(
        best.to_string(index=False)
    )


# ------------------------------------------------------------
# 18I — Project-level economic summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18I — PROJECT-LEVEL ECONOMICS")
print("=" * 60)

if len(conservative):

    project_economics = (
        conservative
        .groupby("PROJECT_EN")
        .agg(
            opportunities=(
                "TRANSACTION_NUMBER",
                "count"
            ),
            mean_return=(
                "forward_return",
                "mean"
            ),
            median_return=(
                "forward_return",
                "median"
            ),
            positive_rate=(
                "forward_return",
                lambda x: (x > 0).mean()
            ),
            above_10_rate=(
                "forward_return",
                lambda x: (x > 0.10).mean()
            )
        )
        .sort_values(
            "opportunities",
            ascending=False
        )
    )

    project_economics[
        "mean_return"
    ] *= 100

    project_economics[
        "median_return"
    ] *= 100

    project_economics[
        "positive_rate"
    ] *= 100

    project_economics[
        "above_10_rate"
    ] *= 100

    print(
        project_economics.to_string()
    )


# ------------------------------------------------------------
# 18J — Final economic summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18J — ECONOMIC SUMMARY")
print("=" * 60)

if len(conservative):

    print(
        "Conservative opportunities:",
        len(conservative)
    )

    print(
        "Mean return:",
        f"{conservative['forward_return'].mean() * 100:.2f}%"
    )

    print(
        "Median return:",
        f"{conservative['forward_return'].median() * 100:.2f}%"
    )

    print(
        "Positive rate:",
        f"{(conservative['forward_return'] > 0).mean() * 100:.2f}%"
    )

    print(
        ">10% rate:",
        f"{(conservative['forward_return'] > 0.10).mean() * 100:.2f}%"
    )

    print(
        "Worst trade:",
        f"{conservative['forward_return'].min() * 100:.2f}%"
    )

    print(
        "Best trade:",
        f"{conservative['forward_return'].max() * 100:.2f}%"
    )

print("\n" + "=" * 60)
print("STEP 18 — COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# STEP 19 — EXECUTION / LEAKAGE / ROBUSTNESS AUDIT
# ============================================================

import numpy as np
import pandas as pd

print("=" * 60)
print("STEP 19 — EXECUTION / LEAKAGE / ROBUSTNESS AUDIT")
print("=" * 60)

# ------------------------------------------------------------
# 19A — Confirm the validated universe
# ------------------------------------------------------------

bt = backtest_30d.copy()

print("\nBacktest shape:", bt.shape)
print("Date range:", bt["target_date"].min(), "->", bt["target_date"].max())

required = [
    "TRANSACTION_NUMBER",
    "target_date",
    "PROJECT_EN",
    "current_ppsf",
    "future_30d_ppsf",
    "future_30d_comps",
    "forward_30d_return_pct",
    "opportunity",
    "opportunity_score",
    "confidence_score",
]

missing = [c for c in required if c not in bt.columns]

print("Missing required columns:", missing)

assert not missing, f"Missing columns: {missing}"

# ------------------------------------------------------------
# 19B — Reconstruct the frozen production signal
# ------------------------------------------------------------

bt["final_signal"] = "NO SIGNAL"

bt.loc[
    (bt["opportunity"] == "HIGH OPPORTUNITY") &
    (bt["confidence_score"] >= 80),
    "final_signal"
] = "CONSERVATIVE OPPORTUNITY"

bt.loc[
    (bt["opportunity"] == "HIGH OPPORTUNITY") &
    (bt["confidence_score"] < 80),
    "final_signal"
] = "BROAD OPPORTUNITY"

print("\nFinal signal counts:")
print(bt["final_signal"].value_counts())

# ------------------------------------------------------------
# 19C — Baseline performance
# ------------------------------------------------------------

conservative = bt[
    bt["final_signal"] == "CONSERVATIVE OPPORTUNITY"
].copy()

print("\n" + "=" * 60)
print("19C — BASELINE CONSERVATIVE SIGNAL")
print("=" * 60)

print("Transactions:", len(conservative))

print(
    conservative["forward_30d_return_pct"].describe(
        percentiles=[.10, .25, .50, .75, .90]
    )
)

print(
    "\nPositive rate:",
    (conservative["forward_30d_return_pct"] > 0).mean() * 100
)

print(
    ">10% rate:",
    (conservative["forward_30d_return_pct"] > 10).mean() * 100
)

# ------------------------------------------------------------
# 19D — Remove extreme winners
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19D — OUTLIER / EXTREME-WINNER ROBUSTNESS")
print("=" * 60)

returns = conservative["forward_30d_return_pct"]

for cap in [25, 50, 75, 100, 150]:
    subset = conservative[returns <= cap]

    print(
        f"\nReturn <= {cap}%"
        f" | n={len(subset)}"
        f" | mean={subset['forward_30d_return_pct'].mean():.2f}%"
        f" | median={subset['forward_30d_return_pct'].median():.2f}%"
        f" | positive={(subset['forward_30d_return_pct'] > 0).mean()*100:.2f}%"
        f" | >10%={(subset['forward_30d_return_pct'] > 10).mean()*100:.2f}%"
    )

# ------------------------------------------------------------
# 19E — Remove top winners
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19E — TOP-WINNER REMOVAL TEST")
print("=" * 60)

ordered = conservative.sort_values(
    "forward_30d_return_pct",
    ascending=False
).copy()

for n_remove in [1, 2, 3, 5, 10]:
    subset = ordered.iloc[n_remove:].copy()

    print(
        f"\nRemove top {n_remove}"
        f" | n={len(subset)}"
        f" | mean={subset['forward_30d_return_pct'].mean():.2f}%"
        f" | median={subset['forward_30d_return_pct'].median():.2f}%"
        f" | positive={(subset['forward_30d_return_pct'] > 0).mean()*100:.2f}%"
        f" | >10%={(subset['forward_30d_return_pct'] > 10).mean()*100:.2f}%"
    )

# ------------------------------------------------------------
# 19F — Transaction-friction sensitivity
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19F — TRANSACTION COST / SLIPPAGE SENSITIVITY")
print("=" * 60)

# IMPORTANT:
# These are hypothetical round-trip friction assumptions.
# They are NOT claims about actual Dubai transaction costs.

for friction in [0, 1, 2, 3, 5, 10]:
    adjusted = (
        conservative["forward_30d_return_pct"] - friction
    )

    print(
        f"\nRound-trip friction = {friction:.1f} pp"
        f" | mean={adjusted.mean():.2f}%"
        f" | median={adjusted.median():.2f}%"
        f" | positive={(adjusted > 0).mean()*100:.2f}%"
        f" | >10%={(adjusted > 10).mean()*100:.2f}%"
    )

# ------------------------------------------------------------
# 19G — Future comparable robustness
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19G — FUTURE COMPARABLE ROBUSTNESS")
print("=" * 60)

for minimum in [1, 3, 5, 10, 20, 30, 50]:
    subset = conservative[
        conservative["future_30d_comps"] >= minimum
    ]

    if len(subset) == 0:
        continue

    print(
        f"\nFuture comps >= {minimum}"
        f" | n={len(subset)}"
        f" | mean={subset['forward_30d_return_pct'].mean():.2f}%"
        f" | median={subset['forward_30d_return_pct'].median():.2f}%"
        f" | positive={(subset['forward_30d_return_pct'] > 0).mean()*100:.2f}%"
        f" | >10%={(subset['forward_30d_return_pct'] > 10).mean()*100:.2f}%"
    )

# ------------------------------------------------------------
# 19H — Project concentration
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19H — PROJECT CONCENTRATION")
print("=" * 60)

project_counts = (
    conservative["PROJECT_EN"]
    .value_counts()
)

print(project_counts)

print(
    "\nLargest project share:",
    project_counts.iloc[0] / len(conservative) * 100,
    "%"
)

# ------------------------------------------------------------
# 19I — 20% project cap
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19I — 20% PROJECT CONCENTRATION CAP")
print("=" * 60)

max_per_project = max(
    1,
    int(np.floor(len(conservative) * 0.20))
)

print("Maximum observations per project:", max_per_project)

capped_parts = []

for project, group in conservative.groupby("PROJECT_EN"):
    group = group.sort_values(
        "opportunity_score",
        ascending=False
    )
    capped_parts.append(
        group.head(max_per_project)
    )

capped = pd.concat(capped_parts, ignore_index=True)

print("Original:", len(conservative))
print("After cap:", len(capped))

print(
    "Mean:",
    capped["forward_30d_return_pct"].mean()
)

print(
    "Median:",
    capped["forward_30d_return_pct"].median()
)

print(
    "Positive:",
    (capped["forward_30d_return_pct"] > 0).mean() * 100
)

print(
    ">10%:",
    (capped["forward_30d_return_pct"] > 10).mean() * 100
)

# ------------------------------------------------------------
# 19J — Score monotonicity
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19J — OPPORTUNITY SCORE MONOTONICITY")
print("=" * 60)

score_bins = pd.qcut(
    conservative["opportunity_score"],
    q=min(5, conservative["opportunity_score"].nunique()),
    duplicates="drop"
)

score_analysis = (
    conservative
    .assign(score_bin=score_bins)
    .groupby("score_bin", observed=True)
    .agg(
        transactions=("forward_30d_return_pct", "size"),
        median_return=("forward_30d_return_pct", "median"),
        mean_return=("forward_30d_return_pct", "mean"),
        positive_rate=(
            "forward_30d_return_pct",
            lambda x: (x > 0).mean() * 100
        ),
        above_10_rate=(
            "forward_30d_return_pct",
            lambda x: (x > 10).mean() * 100
        )
    )
)

print(score_analysis)

# ------------------------------------------------------------
# 19K — Confidence robustness
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19K — CONFIDENCE ROBUSTNESS")
print("=" * 60)

for threshold in [70, 75, 80]:
    subset = conservative[
        conservative["confidence_score"] >= threshold
    ]

    print(
        f"\nConfidence >= {threshold}"
        f" | n={len(subset)}"
        f" | mean={subset['forward_30d_return_pct'].mean():.2f}%"
        f" | median={subset['forward_30d_return_pct'].median():.2f}%"
        f" | positive={(subset['forward_30d_return_pct'] > 0).mean()*100:.2f}%"
        f" | >10%={(subset['forward_30d_return_pct'] > 10).mean()*100:.2f}%"
    )

# ------------------------------------------------------------
# 19L — Signal dependency on RAW DISTRICT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19L — REMOVE DOMINANT PROJECT")
print("=" * 60)

without_raw = conservative[
    conservative["PROJECT_EN"] != "RAW DISTRICT BY IMTIAZ R"
].copy()

print("Remaining observations:", len(without_raw))

print(
    "Mean:",
    without_raw["forward_30d_return_pct"].mean()
)

print(
    "Median:",
    without_raw["forward_30d_return_pct"].median()
)

print(
    "Positive:",
    (without_raw["forward_30d_return_pct"] > 0).mean() * 100
)

print(
    ">10%:",
    (without_raw["forward_30d_return_pct"] > 10).mean() * 100
)

# ------------------------------------------------------------
# 19M — Capital simulation using equal weights
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19M — CAPITAL SENSITIVITY")
print("=" * 60)

for capital in [1_000_000, 5_000_000, 10_000_000]:
    allocation = capital / len(conservative)

    gross_profit = (
        allocation *
        conservative["forward_30d_return_pct"].sum() / 100
    )

    print(
        f"\nCapital AED {capital:,.0f}"
        f" | allocation/trade AED {allocation:,.2f}"
        f" | gross P&L AED {gross_profit:,.2f}"
        f" | return {gross_profit/capital*100:.2f}%"
    )

# ------------------------------------------------------------
# 19N — Final audit summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19N — FINAL AUDIT SUMMARY")
print("=" * 60)

print("Conservative observations:", len(conservative))
print(
    "Mean return:",
    round(conservative["forward_30d_return_pct"].mean(), 2),
    "%"
)
print(
    "Median return:",
    round(conservative["forward_30d_return_pct"].median(), 2),
    "%"
)
print(
    "Positive rate:",
    round(
        (conservative["forward_30d_return_pct"] > 0).mean() * 100,
        2
    ),
    "%"
)
print(
    ">10% rate:",
    round(
        (conservative["forward_30d_return_pct"] > 10).mean() * 100,
        2
    ),
    "%"
)

print("\nSTEP 19 — COMPLETE")
```

```python
# ============================================================
# STEP 20 — DEPLOYABILITY & POSITION-SIZING VALIDATION
# ============================================================

import numpy as np
import pandas as pd

print("=" * 60)
print("STEP 20 — DEPLOYABILITY & POSITION-SIZING VALIDATION")
print("=" * 60)

# ------------------------------------------------------------
# 20A — Start from frozen Step 17 signal
# ------------------------------------------------------------

bt = backtest_30d.copy()

bt["final_signal"] = "NO SIGNAL"

bt.loc[
    (bt["opportunity"] == "HIGH OPPORTUNITY") &
    (bt["confidence_score"] >= 80),
    "final_signal"
] = "CONSERVATIVE OPPORTUNITY"

bt.loc[
    (bt["opportunity"] == "HIGH OPPORTUNITY") &
    (bt["confidence_score"] < 80),
    "final_signal"
] = "BROAD OPPORTUNITY"

conservative = bt[
    bt["final_signal"] == "CONSERVATIVE OPPORTUNITY"
].copy()

print("\nConservative opportunities:", len(conservative))

# ------------------------------------------------------------
# 20B — Attach actual transaction economics
#
# Use xgb_df because it contains the original TRANS_VALUE.
# Match on transaction number + transaction date.
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20B — ATTACH ACTUAL TRANSACTION VALUES")
print("=" * 60)

assert "TRANSACTION_NUMBER" in xgb_df.columns
assert "INSTANCE_DATE" in xgb_df.columns
assert "TRANS_VALUE" in xgb_df.columns

lookup_cols = [
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "AREA_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "TRANS_VALUE",
]

value_lookup = xgb_df[lookup_cols].copy()

value_lookup["INSTANCE_DATE"] = pd.to_datetime(
    value_lookup["INSTANCE_DATE"]
)

conservative["target_date"] = pd.to_datetime(
    conservative["target_date"]
)

# The backtest target_date and source INSTANCE_DATE may differ
# by seconds depending on how the backtest was constructed.
# First attempt exact match.
value_lookup_exact = value_lookup.drop_duplicates(
    subset=["TRANSACTION_NUMBER", "INSTANCE_DATE"],
    keep="first"
)

conservative = conservative.merge(
    value_lookup_exact,
    left_on=["TRANSACTION_NUMBER", "target_date"],
    right_on=["TRANSACTION_NUMBER", "INSTANCE_DATE"],
    how="left",
    suffixes=("", "_source")
)

# ------------------------------------------------------------
# 20C — If exact date matching misses values, use transaction
# number only, but only where that transaction has one unique
# TRANS_VALUE.
# ------------------------------------------------------------

missing_value = conservative["TRANS_VALUE"].isna()

unique_value_lookup = (
    value_lookup
    .groupby("TRANSACTION_NUMBER")["TRANS_VALUE"]
    .agg(["nunique", "first"])
)

unique_value_lookup = unique_value_lookup[
    unique_value_lookup["nunique"] == 1
]["first"]

conservative.loc[missing_value, "TRANS_VALUE"] = (
    conservative.loc[missing_value, "TRANSACTION_NUMBER"]
    .map(unique_value_lookup)
)

print(
    "Missing TRANS_VALUE:",
    conservative["TRANS_VALUE"].isna().sum()
)

assert conservative["TRANS_VALUE"].notna().all(), (
    "Some conservative opportunities could not be matched "
    "to an actual transaction value."
)

# ------------------------------------------------------------
# 20D — Transaction-value distribution
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20D — ACTUAL TICKET SIZE DISTRIBUTION")
print("=" * 60)

print(
    conservative["TRANS_VALUE"].describe(
        percentiles=[.10, .25, .50, .75, .90, .95]
    )
)

print(
    "\nTotal historical ticket value:",
    f"AED {conservative['TRANS_VALUE'].sum():,.2f}"
)

print(
    "Median ticket:",
    f"AED {conservative['TRANS_VALUE'].median():,.2f}"
)

print(
    "Largest ticket:",
    f"AED {conservative['TRANS_VALUE'].max():,.2f}"
)

# ------------------------------------------------------------
# 20E — Equal-weight vs value-weighted signal
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20E — EQUAL-WEIGHT VS VALUE-WEIGHTED")
print("=" * 60)

equal_mean = conservative[
    "forward_30d_return_pct"
].mean()

value_weights = (
    conservative["TRANS_VALUE"] /
    conservative["TRANS_VALUE"].sum()
)

value_weighted_return = (
    conservative["forward_30d_return_pct"] *
    value_weights
).sum()

print(
    f"Equal-weight mean return: {equal_mean:.2f}%"
)

print(
    f"Historical-ticket-value-weighted return: "
    f"{value_weighted_return:.2f}%"
)

# ------------------------------------------------------------
# 20F — Capital feasibility
#
# A historical transaction is treated as one indivisible ticket.
# We ask how many identified opportunities could theoretically
# be purchased if one unit of each transaction were required.
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20F — CAPITAL FEASIBILITY")
print("=" * 60)

for capital in [1_000_000, 5_000_000, 10_000_000]:

    feasible = conservative[
        conservative["TRANS_VALUE"] <= capital
    ].copy()

    print(
        f"\nCapital AED {capital:,.0f}"
    )

    print(
        "Opportunities individually affordable:",
        len(feasible),
        "/",
        len(conservative)
    )

    print(
        "Total ticket value of affordable opportunities:",
        f"AED {feasible['TRANS_VALUE'].sum():,.2f}"
    )

# ------------------------------------------------------------
# 20G — Greedy portfolio construction
#
# Rank by opportunity_score and add whole transaction tickets
# until the capital budget is exhausted.
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20G — SCORE-RANKED CAPITAL DEPLOYMENT")
print("=" * 60)

def build_portfolio(
    data,
    capital,
    max_project_share=None
):
    df = data.sort_values(
        "opportunity_score",
        ascending=False
    ).copy()

    selected = []
    spent = 0.0
    project_spend = {}

    for _, row in df.iterrows():

        value = float(row["TRANS_VALUE"])
        project = row["PROJECT_EN"]

        if spent + value > capital:
            continue

        if max_project_share is not None:
            current_project_spend = project_spend.get(
                project,
                0.0
            )

            if (
                current_project_spend + value
                > capital * max_project_share
            ):
                continue

        selected.append(row)

        spent += value

        project_spend[project] = (
            project_spend.get(project, 0.0) + value
        )

    if not selected:
        return pd.DataFrame()

    return pd.DataFrame(selected)


portfolio_results = {}

for capital in [1_000_000, 5_000_000, 10_000_000]:

    p = build_portfolio(
        conservative,
        capital
    )

    portfolio_results[capital] = p

    print(
        f"\nCapital AED {capital:,.0f}"
    )

    if len(p) == 0:
        print("No complete transaction tickets fit.")
        continue

    invested = p["TRANS_VALUE"].sum()

    gross_profit = (
        p["TRANS_VALUE"] *
        p["forward_30d_return_pct"] / 100
    ).sum()

    print("Transactions:", len(p))
    print("Capital deployed:", f"AED {invested:,.2f}")
    print(
        "Capital utilization:",
        f"{invested / capital * 100:.2f}%"
    )
    print(
        "Gross simulated profit:",
        f"AED {gross_profit:,.2f}"
    )
    print(
        "Return on deployed capital:",
        f"{gross_profit / invested * 100:.2f}%"
    )
    print(
        "Return on total capital:",
        f"{gross_profit / capital * 100:.2f}%"
    )

# ------------------------------------------------------------
# 20H — 10% PROJECT CONCENTRATION CAP
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20H — 10% PROJECT CAPITAL CAP")
print("=" * 60)

for capital in [1_000_000, 5_000_000, 10_000_000]:

    p = build_portfolio(
        conservative,
        capital,
        max_project_share=0.10
    )

    print(
        f"\nCapital AED {capital:,.0f}"
    )

    if len(p) == 0:
        print("No complete transaction tickets fit.")
        continue

    invested = p["TRANS_VALUE"].sum()

    gross_profit = (
        p["TRANS_VALUE"] *
        p["forward_30d_return_pct"] / 100
    ).sum()

    print("Transactions:", len(p))
    print(
        "Capital deployed:",
        f"AED {invested:,.2f}"
    )
    print(
        "Utilization:",
        f"{invested / capital * 100:.2f}%"
    )
    print(
        "Gross profit:",
        f"AED {gross_profit:,.2f}"
    )
    print(
        "Return on deployed:",
        f"{gross_profit / invested * 100:.2f}%"
    )

# ------------------------------------------------------------
# 20I — 20% PROJECT CAPITAL CAP
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20I — 20% PROJECT CAPITAL CAP")
print("=" * 60)

for capital in [1_000_000, 5_000_000, 10_000_000]:

    p = build_portfolio(
        conservative,
        capital,
        max_project_share=0.20
    )

    print(
        f"\nCapital AED {capital:,.0f}"
    )

    if len(p) == 0:
        print("No complete transaction tickets fit.")
        continue

    invested = p["TRANS_VALUE"].sum()

    gross_profit = (
        p["TRANS_VALUE"] *
        p["forward_30d_return_pct"] / 100
    ).sum()

    print("Transactions:", len(p))
    print(
        "Capital deployed:",
        f"AED {invested:,.2f}"
    )
    print(
        "Utilization:",
        f"{invested / capital * 100:.2f}%"
    )
    print(
        "Gross profit:",
        f"AED {gross_profit:,.2f}"
    )
    print(
        "Return on deployed:",
        f"{gross_profit / invested * 100:.2f}%"
    )

# ------------------------------------------------------------
# 20J — RETURN AFTER HYPOTHETICAL FRICTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20J — DEPLOYED PORTFOLIO FRICTION")
print("=" * 60)

# Apply friction to the return percentage itself.
# This remains a sensitivity analysis, not an assertion of
# actual Dubai transaction costs.

for friction in [2, 5, 10, 15, 20]:

    adjusted_return = (
        conservative["forward_30d_return_pct"]
        - friction
    )

    print(
        f"\nFriction = {friction} percentage points"
    )
    print(
        "Mean:",
        f"{adjusted_return.mean():.2f}%"
    )
    print(
        "Median:",
        f"{adjusted_return.median():.2f}%"
    )
    print(
        "Positive:",
        f"{(adjusted_return > 0).mean()*100:.2f}%"
    )
    print(
        ">10%:",
        f"{(adjusted_return > 10).mean()*100:.2f}%"
    )

# ------------------------------------------------------------
# 20K — PROPERTY-TYPE CONCENTRATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20K — PROPERTY-TYPE CONCENTRATION")
print("=" * 60)

print(
    conservative["ROOMS_EN"].value_counts()
)

# ------------------------------------------------------------
# 20L — OFF-PLAN CONCENTRATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20L — OFF-PLAN / READY CONCENTRATION")
print("=" * 60)

print(
    conservative["IS_OFFPLAN_EN"].value_counts()
)

# ------------------------------------------------------------
# 20M — AREA CONCENTRATION
#
# AREA_EN is not guaranteed to be present in backtest_30d.
# Recover it from xgb_df if necessary.
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20M — AREA CONCENTRATION")
print("=" * 60)

if "AREA_EN" not in conservative.columns:

    area_lookup = (
        xgb_df[
            [
                "TRANSACTION_NUMBER",
                "INSTANCE_DATE",
                "AREA_EN"
            ]
        ]
        .drop_duplicates(
            subset=["TRANSACTION_NUMBER", "INSTANCE_DATE"]
        )
        .copy()
    )

    conservative = conservative.merge(
        area_lookup,
        left_on=["TRANSACTION_NUMBER", "target_date"],
        right_on=["TRANSACTION_NUMBER", "INSTANCE_DATE"],
        how="left",
        suffixes=("", "_area")
    )

print(
    conservative["AREA_EN"].value_counts()
)

# ------------------------------------------------------------
# 20N — FINAL DEPLOYABILITY SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("20N — FINAL DEPLOYABILITY SUMMARY")
print("=" * 60)

print(
    "Research opportunities:",
    len(conservative)
)

print(
    "Mean forward comparable return:",
    f"{conservative['forward_30d_return_pct'].mean():.2f}%"
)

print(
    "Median forward comparable return:",
    f"{conservative['forward_30d_return_pct'].median():.2f}%"
)

print(
    "Median transaction value:",
    f"AED {conservative['TRANS_VALUE'].median():,.2f}"
)

print(
    "Total historical transaction value:",
    f"AED {conservative['TRANS_VALUE'].sum():,.2f}"
)

print(
    "Largest project:",
    conservative["PROJECT_EN"].value_counts().index[0]
)

print(
    "Largest project observation share:",
    f"{conservative['PROJECT_EN'].value_counts().iloc[0] / len(conservative) * 100:.2f}%"
)

print(
    "\nSTEP 20 — COMPLETE"
)
```

```python
# ============================================================
# STEP 21 — MISPRICING VS MOMENTUM / PROJECT-ADJUSTED TEST
# ============================================================

import numpy as np
import pandas as pd

print("=" * 60)
print("STEP 21 — MISPRICING VS MOMENTUM / PROJECT-ADJUSTED TEST")
print("=" * 60)

# ------------------------------------------------------------
# 21A — Rebuild frozen conservative signal
# ------------------------------------------------------------

bt = backtest_30d.copy()

bt["final_signal"] = "NO SIGNAL"

bt.loc[
    (bt["opportunity"] == "HIGH OPPORTUNITY") &
    (bt["confidence_score"] >= 80),
    "final_signal"
] = "CONSERVATIVE OPPORTUNITY"

bt.loc[
    (bt["opportunity"] == "HIGH OPPORTUNITY") &
    (bt["confidence_score"] < 80),
    "final_signal"
] = "BROAD OPPORTUNITY"

conservative = bt[
    bt["final_signal"] == "CONSERVATIVE OPPORTUNITY"
].copy()

print("\nConservative opportunities:", len(conservative))

# ------------------------------------------------------------
# 21B — Recover project-level current/future information
# ------------------------------------------------------------
#
# We need to determine whether the signal is merely selecting
# projects that subsequently appreciated strongly.
#
# The backtest already contains transaction-level current/future
# PPSF. We therefore calculate project-level forward performance
# from the same backtest universe.
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21B — PROJECT-LEVEL FORWARD PERFORMANCE")
print("=" * 60)

project_forward = (
    bt.groupby("PROJECT_EN")
    .agg(
        project_transactions=(
            "forward_30d_return_pct",
            "size"
        ),
        project_median_forward_return=(
            "forward_30d_return_pct",
            "median"
        ),
        project_mean_forward_return=(
            "forward_30d_return_pct",
            "mean"
        ),
        project_current_ppsf_median=(
            "current_ppsf",
            "median"
        ),
        project_future_ppsf_median=(
            "future_30d_ppsf",
            "median"
        )
    )
)

print(
    project_forward.sort_values(
        "project_median_forward_return",
        ascending=False
    ).head(20)
)

# ------------------------------------------------------------
# 21C — Attach project forward return to every opportunity
# ------------------------------------------------------------

conservative = conservative.merge(
    project_forward[
        [
            "project_transactions",
            "project_median_forward_return",
            "project_mean_forward_return",
            "project_current_ppsf_median",
            "project_future_ppsf_median"
        ]
    ],
    left_on="PROJECT_EN",
    right_index=True,
    how="left"
)

# ------------------------------------------------------------
# 21D — Opportunity return minus project return
# ------------------------------------------------------------

conservative["project_adjusted_return"] = (
    conservative["forward_30d_return_pct"]
    - conservative["project_median_forward_return"]
)

print("\n" + "=" * 60)
print("21D — PROJECT-ADJUSTED PERFORMANCE")
print("=" * 60)

print(
    conservative[
        "project_adjusted_return"
    ].describe(
        percentiles=[.10, .25, .50, .75, .90]
    )
)

print(
    "\nMean project-adjusted return:",
    f"{conservative['project_adjusted_return'].mean():.2f}%"
)

print(
    "Median project-adjusted return:",
    f"{conservative['project_adjusted_return'].median():.2f}%"
)

print(
    "Positive project-adjusted rate:",
    f"{(conservative['project_adjusted_return'] > 0).mean()*100:.2f}%"
)

# ------------------------------------------------------------
# 21E — Is the signal selecting unusually strong projects?
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21E — SIGNAL VS PROJECT MOMENTUM")
print("=" * 60)

signal_project_stats = (
    conservative
    .groupby("PROJECT_EN")
    .agg(
        signal_count=(
            "forward_30d_return_pct",
            "size"
        ),
        signal_mean_return=(
            "forward_30d_return_pct",
            "mean"
        ),
        signal_median_return=(
            "forward_30d_return_pct",
            "median"
        ),
        project_median_return=(
            "project_median_forward_return",
            "first"
        ),
        project_mean_return=(
            "project_mean_forward_return",
            "first"
        ),
        adjusted_mean=(
            "project_adjusted_return",
            "mean"
        ),
        score_mean=(
            "opportunity_score",
            "mean"
        )
    )
    .sort_values(
        "signal_count",
        ascending=False
    )
)

print(signal_project_stats)

# ------------------------------------------------------------
# 21F — Remove projects with very high underlying momentum
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21F — REMOVE HIGH-MOMENTUM PROJECTS")
print("=" * 60)

for threshold in [10, 20, 30, 40, 50]:

    subset = conservative[
        conservative["project_median_forward_return"]
        < threshold
    ]

    if len(subset) == 0:
        continue

    print(
        f"\nProject median forward return < {threshold}%"
        f" | n={len(subset)}"
        f" | mean={subset['forward_30d_return_pct'].mean():.2f}%"
        f" | median={subset['forward_30d_return_pct'].median():.2f}%"
        f" | positive={(subset['forward_30d_return_pct'] > 0).mean()*100:.2f}%"
        f" | >10%={(subset['forward_30d_return_pct'] > 10).mean()*100:.2f}%"
        f" | project-adjusted mean={subset['project_adjusted_return'].mean():.2f}%"
    )

# ------------------------------------------------------------
# 21G — Within-project comparison
# ------------------------------------------------------------
#
# Compare each signal observation with the distribution of other
# transactions in the SAME PROJECT.
#
# This is a stronger test of whether the selected transaction
# subsequently outperformed its project peers.
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21G — WITHIN-PROJECT SIGNAL TEST")
print("=" * 60)

project_return_distribution = (
    bt.groupby("PROJECT_EN")[
        "forward_30d_return_pct"
    ]
    .agg(
        project_return_mean="mean",
        project_return_median="median",
        project_return_std="std"
    )
)

conservative = conservative.merge(
    project_return_distribution,
    left_on="PROJECT_EN",
    right_index=True,
    how="left"
)

conservative["within_project_excess"] = (
    conservative["forward_30d_return_pct"]
    - conservative["project_return_mean"]
)

print(
    conservative[
        "within_project_excess"
    ].describe(
        percentiles=[.10, .25, .50, .75, .90]
    )
)

print(
    "\nMean within-project excess:",
    f"{conservative['within_project_excess'].mean():.2f}%"
)

print(
    "Median within-project excess:",
    f"{conservative['within_project_excess'].median():.2f}%"
)

print(
    "Positive within-project excess:",
    f"{(conservative['within_project_excess'] > 0).mean()*100:.2f}%"
)

# ------------------------------------------------------------
# 21H — Compare against same-project median
# ------------------------------------------------------------

conservative["excess_vs_project_median"] = (
    conservative["forward_30d_return_pct"]
    - conservative["project_median_forward_return"]
)

print("\n" + "=" * 60)
print("21H — EXCESS VS PROJECT MEDIAN")
print("=" * 60)

print(
    "Mean excess:",
    f"{conservative['excess_vs_project_median'].mean():.2f}%"
)

print(
    "Median excess:",
    f"{conservative['excess_vs_project_median'].median():.2f}%"
)

print(
    "Positive excess rate:",
    f"{(conservative['excess_vs_project_median'] > 0).mean()*100:.2f}%"
)

# ------------------------------------------------------------
# 21I — Score relationship with project-adjusted return
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21I — OPPORTUNITY SCORE VS PROJECT-ADJUSTED RETURN")
print("=" * 60)

try:
    score_corr = conservative[
        [
            "opportunity_score",
            "forward_30d_return_pct",
            "project_adjusted_return",
            "within_project_excess"
        ]
    ].corr(method="spearman")

    print(score_corr)
except Exception as e:
    print("Correlation calculation failed:", e)

# ------------------------------------------------------------
# 21J — Score quintiles against project-adjusted return
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21J — SCORE QUINTILES / PROJECT-ADJUSTED RETURN")
print("=" * 60)

score_bins = pd.qcut(
    conservative["opportunity_score"],
    q=min(
        5,
        conservative["opportunity_score"].nunique()
    ),
    duplicates="drop"
)

score_adjusted = (
    conservative
    .assign(score_bin=score_bins)
    .groupby(
        "score_bin",
        observed=True
    )
    .agg(
        transactions=(
            "project_adjusted_return",
            "size"
        ),
        median_adjusted=(
            "project_adjusted_return",
            "median"
        ),
        mean_adjusted=(
            "project_adjusted_return",
            "mean"
        ),
        positive_adjusted=(
            "project_adjusted_return",
            lambda x: (x > 0).mean() * 100
        )
    )
)

print(score_adjusted)

# ------------------------------------------------------------
# 21K — Remove dominant project
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21K — PROJECT-ADJUSTED RESULT WITHOUT RAW DISTRICT")
print("=" * 60)

without_raw = conservative[
    conservative["PROJECT_EN"] !=
    "RAW DISTRICT BY IMTIAZ R"
].copy()

print("Observations:", len(without_raw))

print(
    "Mean raw return:",
    f"{without_raw['forward_30d_return_pct'].mean():.2f}%"
)

print(
    "Mean project-adjusted return:",
    f"{without_raw['project_adjusted_return'].mean():.2f}%"
)

print(
    "Median project-adjusted return:",
    f"{without_raw['project_adjusted_return'].median():.2f}%"
)

print(
    "Positive project-adjusted:",
    f"{(without_raw['project_adjusted_return'] > 0).mean()*100:.2f}%"
)

# ------------------------------------------------------------
# 21L — Conservative signal against NON-HIGH project-adjusted
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21L — SIGNAL VS NON-SIGNAL WITHIN PROJECT")
print("=" * 60)

# Calculate each transaction's project median excluding itself.
# This prevents the signal transaction from influencing its own
# benchmark.

project_stats_excl_self = (
    bt.groupby("PROJECT_EN")[
        "forward_30d_return_pct"
    ]
    .agg(
        project_sum="sum",
        project_count="count"
    )
)

conservative["project_peer_mean_excl_self"] = (
    conservative["PROJECT_EN"]
    .map(project_stats_excl_self["project_sum"])
    -
    conservative["forward_30d_return_pct"]
)

conservative["project_peer_count_excl_self"] = (
    conservative["PROJECT_EN"]
    .map(project_stats_excl_self["project_count"])
    - 1
)

conservative["peer_adjusted_return"] = np.where(
    conservative["project_peer_count_excl_self"] > 0,
    conservative["forward_30d_return_pct"]
    -
    (
        conservative["project_peer_mean_excl_self"]
        /
        conservative["project_peer_count_excl_self"]
    ),
    np.nan
)

peer_valid = conservative[
    conservative["peer_adjusted_return"].notna()
].copy()

print(
    "Signal observations with project peers:",
    len(peer_valid)
)

print(
    "Mean peer-adjusted return:",
    f"{peer_valid['peer_adjusted_return'].mean():.2f}%"
)

print(
    "Median peer-adjusted return:",
    f"{peer_valid['peer_adjusted_return'].median():.2f}%"
)

print(
    "Positive peer-adjusted rate:",
    f"{(peer_valid['peer_adjusted_return'] > 0).mean()*100:.2f}%"
)

# ------------------------------------------------------------
# 21M — FINAL CLASSIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("21M — FINAL SIGNAL CLASSIFICATION")
print("=" * 60)

raw_mean = conservative[
    "forward_30d_return_pct"
].mean()

adjusted_mean = conservative[
    "project_adjusted_return"
].mean()

adjusted_positive = (
    conservative["project_adjusted_return"] > 0
).mean() * 100

print(
    f"\nRaw signal mean return: {raw_mean:.2f}%"
)

print(
    f"Project-adjusted mean return: {adjusted_mean:.2f}%"
)

print(
    f"Project-adjusted positive rate: "
    f"{adjusted_positive:.2f}%"
)

if adjusted_mean > 10 and adjusted_positive >= 70:
    classification = "MIXED / STRONG WITHIN-PROJECT EFFECT"

elif adjusted_mean > 5 and adjusted_positive >= 60:
    classification = "MIXED"

elif adjusted_mean > 0:
    classification = "MOMENTUM-DOMINANT"

else:
    classification = "INSUFFICIENT EVIDENCE OF MISPRICING"

print(
    "\nPRELIMINARY CLASSIFICATION:",
    classification
)

print("\nSTEP 21 — COMPLETE")
```

```python
# ============================================================
# STEP 22 — LIVE OPPORTUNITY SCANNER
# ============================================================

import numpy as np
import pandas as pd

print("=" * 60)
print("STEP 22 — LIVE OPPORTUNITY SCANNER")
print("=" * 60)

# ------------------------------------------------------------
# 22A — IDENTIFY THE CURRENT RESEARCH DATAFRAME
# ------------------------------------------------------------

# We need the dataframe that already contains the validated
# opportunity fields. Prefer high_opp / top_clean / arb etc.
# without rebuilding the model.

candidate_names = [
    "high_opp",
    "top_clean",
    "clean",
    "arb",
    "arbitrage_df",
]

available_candidates = []

for name in candidate_names:
    if name in globals():
        obj = globals()[name]
        if isinstance(obj, pd.DataFrame):
            available_candidates.append(
                (name, obj.shape, list(obj.columns))
            )

print("\nAvailable candidate dataframes:")

for name, shape, cols in available_candidates:
    print(f"{name}: {shape}")

# ------------------------------------------------------------
# 22B — FIND DATAFRAME WITH PRODUCTION SIGNAL FIELDS
# ------------------------------------------------------------

signal_fields = [
    "opportunity",
    "confidence_score",
    "risk_adjusted_gap_pct",
]

source_name = None
source_df = None

for name, shape, cols in available_candidates:
    if all(c in cols for c in signal_fields):
        source_name = name
        source_df = globals()[name].copy()
        break

if source_df is None:
    raise ValueError(
        "No existing dataframe contains the validated "
        "opportunity signal fields."
    )

print("\nSelected source dataframe:")
print(source_name)
print(source_df.shape)

# ------------------------------------------------------------
# 22C — REQUIRED TRANSACTION FIELDS
# ------------------------------------------------------------

required_core = [
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "PROJECT_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "TRANS_VALUE",
    "ACTUAL_AREA",
    "price_per_sqft",
    "opportunity",
    "confidence_score",
    "risk_adjusted_gap_pct",
]

missing_core = [
    c for c in required_core
    if c not in source_df.columns
]

print("\nMissing core fields:")
print(missing_core)

assert not missing_core, (
    f"Production signal source is missing: {missing_core}"
)

# ------------------------------------------------------------
# 22D — FREEZE PRODUCTION RULE
# ------------------------------------------------------------

# DO NOT change these thresholds.

source_df["final_signal"] = "NO SIGNAL"

source_df.loc[
    (
        source_df["opportunity"].eq("HIGH OPPORTUNITY")
        & (source_df["confidence_score"] >= 80)
    ),
    "final_signal"
] = "CONSERVATIVE OPPORTUNITY"

source_df.loc[
    (
        source_df["opportunity"].eq("HIGH OPPORTUNITY")
        & (source_df["confidence_score"] < 80)
    ),
    "final_signal"
] = "BROAD OPPORTUNITY"

# ------------------------------------------------------------
# 22E — CALCULATE / VERIFY PPSF
# ------------------------------------------------------------

source_df["calculated_ppsf"] = (
    source_df["TRANS_VALUE"]
    / source_df["ACTUAL_AREA"].replace(0, np.nan)
)

# Compare with existing PPSF where available
source_df["ppsf_difference"] = (
    source_df["price_per_sqft"]
    - source_df["calculated_ppsf"]
)

print("\nPPSF consistency:")
print(
    source_df["ppsf_difference"]
    .abs()
    .describe()
)

# ------------------------------------------------------------
# 22F — OPPORTUNITY SCORE CHECK
# ------------------------------------------------------------

if "opportunity_score" in source_df.columns:

    score_summary = (
        source_df.loc[
            source_df["final_signal"] != "NO SIGNAL",
            [
                "opportunity_score",
                "confidence_score",
                "risk_adjusted_gap_pct",
            ]
        ]
        .describe()
    )

    print("\nValidated signal score summary:")
    print(score_summary)

else:

    print(
        "\nWARNING: opportunity_score is not present "
        "in this source dataframe."
    )

# ------------------------------------------------------------
# 22G — FINAL LIVE SCANNER OUTPUT
# ------------------------------------------------------------

scanner_cols = [
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "PROJECT_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "TRANS_VALUE",
    "ACTUAL_AREA",
    "price_per_sqft",
    "fair_value",
    "valuation_gap_aed",
    "valuation_gap_pct",
    "risk_adjusted_gap_pct",
    "confidence_score",
    "opportunity",
]

if "opportunity_score" in source_df.columns:
    scanner_cols.append("opportunity_score")

scanner_cols.append("final_signal")

scanner_cols = [
    c for c in scanner_cols
    if c in source_df.columns
]

live_scanner = (
    source_df[scanner_cols]
    .copy()
    .sort_values(
        [
            "final_signal",
            "confidence_score",
            "risk_adjusted_gap_pct",
        ],
        ascending=[True, False, False]
    )
)

# ------------------------------------------------------------
# 22H — SIGNAL COUNTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("22H — LIVE SIGNAL COUNTS")
print("=" * 60)

print(
    live_scanner["final_signal"]
    .value_counts()
)

# ------------------------------------------------------------
# 22I — CONSERVATIVE OPPORTUNITIES
# ------------------------------------------------------------

conservative_live = live_scanner[
    live_scanner["final_signal"]
    == "CONSERVATIVE OPPORTUNITY"
].copy()

broad_live = live_scanner[
    live_scanner["final_signal"]
    == "BROAD OPPORTUNITY"
].copy()

print("\nConservative opportunities:")
print(len(conservative_live))

print("\nBroad opportunities:")
print(len(broad_live))

# ------------------------------------------------------------
# 22J — DISPLAY CURRENT OPPORTUNITIES
# ------------------------------------------------------------

if len(conservative_live) > 0:

    display_cols = [
        "TRANSACTION_NUMBER",
        "INSTANCE_DATE",
        "PROJECT_EN",
        "ROOMS_EN",
        "IS_OFFPLAN_EN",
        "TRANS_VALUE",
        "price_per_sqft",
        "fair_value",
        "risk_adjusted_gap_pct",
        "confidence_score",
    ]

    if "opportunity_score" in conservative_live.columns:
        display_cols.append("opportunity_score")

    print("\n" + "=" * 60)
    print("22J — CONSERVATIVE OPPORTUNITIES")
    print("=" * 60)

    print(
        conservative_live[
            [
                c for c in display_cols
                if c in conservative_live.columns
            ]
        ].to_string(index=False)
    )

else:

    print(
        "\nNo Conservative Opportunities found "
        "in the current dataframe."
    )

# ------------------------------------------------------------
# 22K — PRODUCTION VALIDATION CHECKS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("22K — PRODUCTION VALIDATION")
print("=" * 60)

print(
    "Conservative count:",
    len(conservative_live)
)

print(
    "Broad count:",
    len(broad_live)
)

print(
    "No signal count:",
    (
        live_scanner["final_signal"]
        == "NO SIGNAL"
    ).sum()
)

# Conservative rule integrity
if len(conservative_live) > 0:

    assert (
        conservative_live["opportunity"]
        .eq("HIGH OPPORTUNITY")
        .all()
    )

    assert (
        conservative_live["confidence_score"]
        >= 80
    ).all()

# Broad rule integrity
if len(broad_live) > 0:

    assert (
        broad_live["opportunity"]
        .eq("HIGH OPPORTUNITY")
        .all()
    )

    assert (
        broad_live["confidence_score"]
        < 80
    ).all()

print("\nProduction rule integrity: PASSED")

# ------------------------------------------------------------
# 22L — FREEZE SPECIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("22L — FROZEN PRODUCTION SPECIFICATION")
print("=" * 60)

print("""
CONSERVATIVE OPPORTUNITY
    opportunity == HIGH OPPORTUNITY
    AND
    confidence_score >= 80

BROAD OPPORTUNITY
    opportunity == HIGH OPPORTUNITY
    AND
    confidence_score < 80

NO SIGNAL
    Everything else

IMPORTANT:
    No XGBoost reconstruction.
    No future information.
    No threshold optimization.
    No modification of the validated signal.
""")

print("=" * 60)
print("STEP 22 — COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# STEP 24 — CURRENT OPPORTUNITY RANKING & CAPITAL ALLOCATION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 60)
print("STEP 24 — CURRENT OPPORTUNITY RANKING & CAPITAL ALLOCATION")
print("=" * 60)

# ------------------------------------------------------------
# 24A — USE THE FROZEN PRODUCTION SOURCE
# ------------------------------------------------------------

prod = high_opp.copy()

prod["INSTANCE_DATE"] = pd.to_datetime(
    prod["INSTANCE_DATE"], errors="coerce"
)

prod["TRANS_VALUE"] = pd.to_numeric(
    prod["TRANS_VALUE"], errors="coerce"
)

prod["opportunity_score"] = pd.to_numeric(
    prod["opportunity_score"], errors="coerce"
)

prod["confidence_score"] = pd.to_numeric(
    prod["confidence_score"], errors="coerce"
)

prod["risk_adjusted_gap_pct"] = pd.to_numeric(
    prod["risk_adjusted_gap_pct"], errors="coerce"
)

# ------------------------------------------------------------
# 24B — APPLY ONLY THE FROZEN PRODUCTION RULE
# ------------------------------------------------------------

prod["final_signal"] = np.select(
    [
        (prod["opportunity"] == "HIGH OPPORTUNITY")
        & (prod["confidence_score"] >= 80),

        (prod["opportunity"] == "HIGH OPPORTUNITY")
        & (prod["confidence_score"] < 80),
    ],
    [
        "CONSERVATIVE OPPORTUNITY",
        "BROAD OPPORTUNITY",
    ],
    default="NO SIGNAL"
)

print("\nSignal counts:")
print(prod["final_signal"].value_counts())

# ------------------------------------------------------------
# 24C — CONSERVATIVE OPPORTUNITIES
# ------------------------------------------------------------

conservative = prod[
    prod["final_signal"] == "CONSERVATIVE OPPORTUNITY"
].copy()

# Ranking is descriptive only:
# 1. opportunity score
# 2. risk-adjusted gap
# 3. confidence
#
# NO new threshold is introduced.

conservative = conservative.sort_values(
    [
        "opportunity_score",
        "risk_adjusted_gap_pct",
        "confidence_score",
    ],
    ascending=[False, False, False]
).reset_index(drop=True)

conservative["rank"] = np.arange(
    1, len(conservative) + 1
)

print("\n" + "=" * 60)
print("24C — CONSERVATIVE OPPORTUNITIES")
print("=" * 60)

display_cols = [
    "rank",
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "PROJECT_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "TRANS_VALUE",
    "price_per_sqft",
    "fair_value",
    "risk_adjusted_gap_pct",
    "confidence_score",
    "opportunity_score",
]

display(
    conservative[display_cols].to_string(index=False)
)

# ------------------------------------------------------------
# 24D — BROAD OPPORTUNITIES
# ------------------------------------------------------------

broad = prod[
    prod["final_signal"] == "BROAD OPPORTUNITY"
].copy()

broad = broad.sort_values(
    [
        "opportunity_score",
        "risk_adjusted_gap_pct",
        "confidence_score",
    ],
    ascending=[False, False, False]
).reset_index(drop=True)

broad["rank"] = np.arange(
    1, len(broad) + 1
)

print("\n" + "=" * 60)
print("24D — BROAD OPPORTUNITIES")
print("=" * 60)

display(
    broad[display_cols].to_string(index=False)
)

# ------------------------------------------------------------
# 24E — CONSERVATIVE SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("24E — CONSERVATIVE SIGNAL SUMMARY")
print("=" * 60)

print("Conservative opportunities:", len(conservative))
print(
    "Total ticket value: AED {:,.2f}".format(
        conservative["TRANS_VALUE"].sum()
    )
)
print(
    "Median ticket: AED {:,.2f}".format(
        conservative["TRANS_VALUE"].median()
    )
)
print(
    "Largest ticket: AED {:,.2f}".format(
        conservative["TRANS_VALUE"].max()
    )
)

print("\nProject concentration:")
print(
    conservative["PROJECT_EN"]
    .value_counts()
)

# ------------------------------------------------------------
# 24F — CAPITAL DEPLOYMENT FUNCTION
# ------------------------------------------------------------

def deploy_by_score(
    opportunities,
    capital,
    project_cap_pct=None
):
    """
    Rank opportunities by the frozen signal score and select
    complete transaction tickets that fit within available capital.

    project_cap_pct:
        Optional maximum percentage of total capital allocated
        to any one project.

    No prediction of future return is used here.
    """

    df = opportunities.copy()

    df = df.sort_values(
        [
            "opportunity_score",
            "risk_adjusted_gap_pct",
        ],
        ascending=[False, False]
    ).reset_index(drop=True)

    selected = []
    deployed = 0.0
    project_deployed = {}

    for _, row in df.iterrows():

        ticket = float(row["TRANS_VALUE"])
        project = row["PROJECT_EN"]

        if not np.isfinite(ticket) or ticket <= 0:
            continue

        # Complete ticket must fit.
        if deployed + ticket > capital:
            continue

        # Optional project-level capital cap.
        if project_cap_pct is not None:

            max_project_capital = (
                capital * project_cap_pct
            )

            already_project = project_deployed.get(
                project, 0.0
            )

            if (
                already_project + ticket
                > max_project_capital
            ):
                continue

        selected.append(row)

        deployed += ticket

        project_deployed[project] = (
            project_deployed.get(project, 0.0)
            + ticket
        )

    result = pd.DataFrame(selected)

    if len(result) == 0:
        return result, deployed

    return result, deployed


# ------------------------------------------------------------
# 24G — CAPITAL SCENARIOS
# ------------------------------------------------------------

capital_scenarios = [
    1_000_000,
    5_000_000,
    10_000_000,
]

print("\n" + "=" * 60)
print("24G — SCORE-RANKED CAPITAL DEPLOYMENT")
print("=" * 60)

deployment_results = {}

for capital in capital_scenarios:

    selected, deployed = deploy_by_score(
        conservative,
        capital
    )

    deployment_results[capital] = selected

    print("\nCapital: AED {:,.0f}".format(capital))
    print(
        "Transactions:",
        len(selected)
    )
    print(
        "Capital deployed: AED {:,.2f}".format(
            deployed
        )
    )
    print(
        "Capital utilization: {:.2f}%".format(
            100 * deployed / capital
        )
    )

    if len(selected):
        print("\nSelected projects:")
        print(
            selected[
                [
                    "PROJECT_EN",
                    "TRANS_VALUE",
                    "opportunity_score",
                    "risk_adjusted_gap_pct",
                ]
            ].to_string(index=False)
        )

# ------------------------------------------------------------
# 24H — 20% PROJECT CAPITAL CAP
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("24H — 20% PROJECT CAPITAL CAP")
print("=" * 60)

for capital in capital_scenarios:

    selected, deployed = deploy_by_score(
        conservative,
        capital,
        project_cap_pct=0.20
    )

    print(
        "\nCapital: AED {:,.0f}".format(capital)
    )
    print(
        "Transactions:",
        len(selected)
    )
    print(
        "Capital deployed: AED {:,.2f}".format(
            deployed
        )
    )
    print(
        "Utilization: {:.2f}%".format(
            100 * deployed / capital
        )
    )

    if len(selected):

        print("\nProject allocation:")

        allocation = (
            selected
            .groupby("PROJECT_EN")["TRANS_VALUE"]
            .sum()
            .sort_values(ascending=False)
        )

        allocation_pct = (
            allocation / capital * 100
        )

        print(
            pd.DataFrame({
                "capital": allocation,
                "capital_pct": allocation_pct
            }).to_string()
        )

# ------------------------------------------------------------
# 24I — CURRENT TOP 10
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("24I — TOP 10 CURRENT CONSERVATIVE OPPORTUNITIES")
print("=" * 60)

top10 = conservative.head(10).copy()

print(
    top10[
        [
            "rank",
            "PROJECT_EN",
            "ROOMS_EN",
            "TRANS_VALUE",
            "price_per_sqft",
            "fair_value",
            "risk_adjusted_gap_pct",
            "opportunity_score",
            "confidence_score",
        ]
    ].to_string(index=False)
)

# ------------------------------------------------------------
# 24J — FINAL VALIDATION
# ------------------------------------------------------------

assert len(conservative) == 44
assert len(broad) == 47
assert len(prod) == 91

print("\n" + "=" * 60)
print("STEP 24 — COMPLETE")
print("=" * 60)

print("Current production candidates:", len(prod))
print("Conservative:", len(conservative))
print("Broad:", len(broad))
print("Production rule unchanged: PASSED")
```

```python
# ============================================================
# STEP 25 — DEAL-LEVEL DUE DILIGENCE & INVESTMENT MEMO
# ============================================================

import pandas as pd
import numpy as np

print("=" * 60)
print("STEP 25 — DEAL-LEVEL DUE DILIGENCE & INVESTMENT MEMO")
print("=" * 60)

# ------------------------------------------------------------
# 25A — SOURCE
# ------------------------------------------------------------

deals = conservative.copy()

required = [
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "PROJECT_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "TRANS_VALUE",
    "price_per_sqft",
    "fair_value",
    "risk_adjusted_gap_pct",
    "confidence_score",
    "opportunity_score",
]

missing = [c for c in required if c not in deals.columns]

print("\nMissing required columns:")
print(missing)

assert not missing, f"Missing required columns: {missing}"

# ------------------------------------------------------------
# 25B — CALCULATE CORE DEAL METRICS
# ------------------------------------------------------------

deals["TRANS_VALUE"] = pd.to_numeric(
    deals["TRANS_VALUE"],
    errors="coerce"
)

deals["fair_value"] = pd.to_numeric(
    deals["fair_value"],
    errors="coerce"
)

deals["price_per_sqft"] = pd.to_numeric(
    deals["price_per_sqft"],
    errors="coerce"
)

deals["risk_adjusted_gap_pct"] = pd.to_numeric(
    deals["risk_adjusted_gap_pct"],
    errors="coerce"
)

deals["opportunity_score"] = pd.to_numeric(
    deals["opportunity_score"],
    errors="coerce"
)

deals["confidence_score"] = pd.to_numeric(
    deals["confidence_score"],
    errors="coerce"
)

deals["discount_to_fair_value_pct"] = (
    1
    - deals["TRANS_VALUE"] / deals["fair_value"]
) * 100

deals["implied_upside_to_fair_value_pct"] = (
    deals["fair_value"] / deals["TRANS_VALUE"] - 1
) * 100

deals["model_value_gap_aed"] = (
    deals["fair_value"] - deals["TRANS_VALUE"]
)

# ------------------------------------------------------------
# 25C — DEAL PRIORITY
# ------------------------------------------------------------

# IMPORTANT:
# This is a REVIEW priority, NOT a new predictive threshold.

deals["deal_priority"] = pd.cut(
    deals["opportunity_score"],
    bins=[-np.inf, 0.65, 0.70, np.inf],
    labels=[
        "STANDARD REVIEW",
        "HIGH PRIORITY",
        "TOP PRIORITY",
    ]
)

deals = deals.sort_values(
    [
        "opportunity_score",
        "risk_adjusted_gap_pct",
        "confidence_score",
    ],
    ascending=[False, False, False]
).reset_index(drop=True)

deals["deal_rank"] = np.arange(
    1,
    len(deals) + 1
)

# ------------------------------------------------------------
# 25D — RED-FLAG SCREEN
# ------------------------------------------------------------

# These are diligence flags only.
# They do NOT invalidate the production signal.

deals["review_flags"] = ""

def add_flag(mask, text):
    deals.loc[mask, "review_flags"] = (
        deals.loc[mask, "review_flags"]
        .astype(str)
        .where(
            deals.loc[mask, "review_flags"].astype(str).str.len() > 0,
            ""
        )
        + np.where(
            deals.loc[mask, "review_flags"].astype(str).str.len() > 0,
            "; ",
            ""
        )
        + text
    )

# Large ticket
add_flag(
    deals["TRANS_VALUE"] >= 1_500_000,
    "LARGE TICKET"
)

# Very high model gap
add_flag(
    deals["risk_adjusted_gap_pct"] >= 30,
    "VERIFY EXTREME GAP"
)

# Lower score within Conservative universe
add_flag(
    deals["opportunity_score"] < 0.65,
    "LOWER SCORE"
)

# ------------------------------------------------------------
# 25E — TOP 10 DEAL MEMO
# ------------------------------------------------------------

top10 = deals.head(10).copy()

memo_cols = [
    "deal_rank",
    "deal_priority",
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "PROJECT_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "TRANS_VALUE",
    "fair_value",
    "model_value_gap_aed",
    "discount_to_fair_value_pct",
    "implied_upside_to_fair_value_pct",
    "price_per_sqft",
    "risk_adjusted_gap_pct",
    "confidence_score",
    "opportunity_score",
    "review_flags",
]

print("\n" + "=" * 60)
print("25E — TOP 10 DEAL MEMO")
print("=" * 60)

print(
    top10[memo_cols].to_string(index=False)
)

# ------------------------------------------------------------
# 25F — PROJECT-LEVEL DUE DILIGENCE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("25F — PROJECT-LEVEL DUE DILIGENCE QUEUE")
print("=" * 60)

project_summary = (
    deals
    .groupby("PROJECT_EN")
    .agg(
        opportunities=("TRANSACTION_NUMBER", "count"),
        total_ticket_value=("TRANS_VALUE", "sum"),
        median_ticket=("TRANS_VALUE", "median"),
        median_gap=("risk_adjusted_gap_pct", "median"),
        median_score=("opportunity_score", "median"),
        max_score=("opportunity_score", "max"),
    )
    .sort_values(
        ["max_score", "median_gap"],
        ascending=False
    )
)

print(
    project_summary.to_string()
)

# ------------------------------------------------------------
# 25G — CAPITAL / DUE-DILIGENCE QUEUE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("25G — CAPITAL PRIORITY QUEUE")
print("=" * 60)

queue = deals[
    [
        "deal_rank",
        "PROJECT_EN",
        "ROOMS_EN",
        "TRANS_VALUE",
        "opportunity_score",
        "risk_adjusted_gap_pct",
        "confidence_score",
        "deal_priority",
        "review_flags",
    ]
].copy()

print(
    queue.head(20).to_string(index=False)
)

# ------------------------------------------------------------
# 25H — MODEL-BASED ECONOMIC SCREEN
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("25H — MODEL ECONOMIC SCREEN")
print("=" * 60)

print(
    "Total Conservative deals:",
    len(deals)
)

print(
    "Total ticket value: AED {:,.2f}".format(
        deals["TRANS_VALUE"].sum()
    )
)

print(
    "Median ticket: AED {:,.2f}".format(
        deals["TRANS_VALUE"].median()
    )
)

print(
    "Median model gap: {:.2f}%".format(
        deals["risk_adjusted_gap_pct"].median()
    )
)

print(
    "Median opportunity score: {:.4f}".format(
        deals["opportunity_score"].median()
    )
)

print(
    "Median implied upside to model fair value: {:.2f}%".format(
        deals["implied_upside_to_fair_value_pct"].median()
    )
)

# ------------------------------------------------------------
# 25I — FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("25I — IMPORTANT INVESTMENT STATUS")
print("=" * 60)

print("""
BUY / REVIEW / PASS classifications are NOT generated
automatically by the model.

The production model identifies candidates.

Final investment approval requires external/property-level
due diligence, including verification of the transaction,
unit, title/project status, liquidity, fees, developer risk,
and whether the model's fair-value comparison is economically
comparable.

Therefore the current model status is:

CONSERVATIVE OPPORTUNITY -> DUE DILIGENCE REQUIRED
BROAD OPPORTUNITY        -> SECONDARY REVIEW
NO SIGNAL                -> DO NOT PRIORITIZE
""")

# ------------------------------------------------------------
# 25J — INTEGRITY CHECK
# ------------------------------------------------------------

assert len(deals) == 44
assert (deals["confidence_score"] >= 80).all()
assert (deals["opportunity"] == "HIGH OPPORTUNITY").all()

print("=" * 60)
print("STEP 25 — COMPLETE")
print("=" * 60)

print("Conservative deals:", len(deals))
print("Production rule unchanged: PASSED")
```

```python
# ============================================================
# EXPORT PRODUCTION DATA — FROZEN SIGNAL DATA FOR STREAMLIT
# ============================================================
#
# PURPOSE:
#   Create the clean, authoritative datasets that Streamlit will use.
#
# OUTPUT:
#   dashboard/data/production_candidates.parquet
#   dashboard/data/backtest_30d.parquet
#   dashboard/data/metadata.json
#
# IMPORTANT:
#   This cell does NOT retrain models.
#   This cell does NOT rebuild the backtest.
#   This cell does NOT change the production rules.
# ============================================================

from pathlib import Path
import json
import pandas as pd
import numpy as np


# ============================================================
# 1. OUTPUT DIRECTORY
# ============================================================

OUTPUT_DIR = Path("dashboard") / "data"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Export directory:")
print(OUTPUT_DIR.resolve())


# ============================================================
# 2. IDENTIFY AUTHORITATIVE DATAFRAMES
# ============================================================

# CURRENT PRODUCTION DATA
#
# Based on the current notebook state, the authoritative current
# production candidate dataframe is expected to be `high_opp`.
#
# We deliberately do NOT use `clean` as the production export.
#
# If your current 91-row production dataframe has a different
# variable name, change ONLY this line.

production_source = high_opp.copy()


# HISTORICAL BACKTEST DATA
#
# This is the validated 30-day backtest dataframe.

backtest_source = backtest_30d.copy()


print("\nSource datasets:")
print("production_source:", production_source.shape)
print("backtest_source:", backtest_source.shape)


# ============================================================
# 3. FREEZE PRODUCTION SIGNAL RULE
# ============================================================

CONSERVATIVE_RULE = (
    (production_source["opportunity"].astype(str).str.upper()
     == "HIGH OPPORTUNITY")
    &
    (pd.to_numeric(
        production_source["confidence_score"],
        errors="coerce"
    ) >= 80)
)

BROAD_RULE = (
    (production_source["opportunity"].astype(str).str.upper()
     == "HIGH OPPORTUNITY")
    &
    (pd.to_numeric(
        production_source["confidence_score"],
        errors="coerce"
    ) < 80)
)


# ============================================================
# 4. CREATE FINAL_SIGNAL
# ============================================================

production_source["final_signal"] = np.select(
    [
        CONSERVATIVE_RULE,
        BROAD_RULE
    ],
    [
        "CONSERVATIVE OPPORTUNITY",
        "BROAD OPPORTUNITY"
    ],
    default="NO SIGNAL"
)


# ============================================================
# 5. NORMALIZE DATES
# ============================================================

if "target_date" in production_source.columns:
    production_source["target_date"] = pd.to_datetime(
        production_source["target_date"],
        errors="coerce"
    )

if "INSTANCE_DATE" in production_source.columns:
    production_source["INSTANCE_DATE"] = pd.to_datetime(
        production_source["INSTANCE_DATE"],
        errors="coerce"
    )

backtest_source["target_date"] = pd.to_datetime(
    backtest_source["target_date"],
    errors="coerce"
)


# ============================================================
# 6. REQUIRED PRODUCTION COLUMNS
# ============================================================

production_required = [
    "TRANSACTION_NUMBER",
    "PROJECT_EN",
    "AREA_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "TRANS_VALUE",
    "price_per_sqft",
    "fair_value",
    "valuation_gap_aed",
    "valuation_gap_pct",
    "risk_adjusted_gap_pct",
    "opportunity_score",
    "confidence_score",
    "opportunity",
    "final_signal",
]


# Some current production versions may not contain every optional
# valuation field. We distinguish required fields from optional ones.

production_optional = [
    "target_date",
    "INSTANCE_DATE",
    "PROP_TYPE_EN",
    "PROP_SB_TYPE_EN",
    "current_ppsf",
    "future_30d_ppsf",
    "future_30d_comps",
    "has_30d_signal",
    "data_quality",
    "model_risk_pct",
    "fair_value_ratio",
    "valuation_gap_pct_capped",
    "opportunity_group",
    "is_high_opportunity",
]


missing_production = [
    c for c in production_required
    if c not in production_source.columns
]

print("\nMissing required production columns:")
print(missing_production)

assert not missing_production, (
    f"Production export cannot continue. "
    f"Missing required columns: {missing_production}"
)


# ============================================================
# 7. REQUIRED BACKTEST COLUMNS
# ============================================================

backtest_required = [
    "TRANSACTION_NUMBER",
    "target_date",
    "PROJECT_EN",
    "ROOMS_EN",
    "IS_OFFPLAN_EN",
    "current_ppsf",
    "future_30d_ppsf",
    "future_30d_comps",
    "has_30d_signal",
    "forward_30d_return_pct",
    "opportunity",
    "opportunity_score",
    "risk_adjusted_gap_pct",
    "confidence_score",
    "opportunity_group",
    "is_high_opportunity",
]


missing_backtest = [
    c for c in backtest_required
    if c not in backtest_source.columns
]

print("\nMissing required backtest columns:")
print(missing_backtest)

assert not missing_backtest, (
    f"Backtest export cannot continue. "
    f"Missing required columns: {missing_backtest}"
)


# ============================================================
# 8. ADD FINAL_SIGNAL TO BACKTEST
# ============================================================

backtest_source["final_signal"] = np.select(
    [
        (
            backtest_source["opportunity"]
            .astype(str)
            .str.upper()
            .eq("HIGH OPPORTUNITY")
            &
            (
                pd.to_numeric(
                    backtest_source["confidence_score"],
                    errors="coerce"
                ) >= 80
            )
        ),
        (
            backtest_source["opportunity"]
            .astype(str)
            .str.upper()
            .eq("HIGH OPPORTUNITY")
            &
            (
                pd.to_numeric(
                    backtest_source["confidence_score"],
                    errors="coerce"
                ) < 80
            )
        ),
    ],
    [
        "CONSERVATIVE OPPORTUNITY",
        "BROAD OPPORTUNITY",
    ],
    default="NO SIGNAL",
)


# ============================================================
# 9. KEEP ONLY USEFUL PRODUCTION COLUMNS
# ============================================================

production_columns = [
    c for c in production_required + production_optional
    if c in production_source.columns
]

production_export = production_source[
    production_columns
].copy()


# ============================================================
# 10. KEEP ONLY USEFUL BACKTEST COLUMNS
# ============================================================

backtest_columns = backtest_required + ["final_signal"]

backtest_export = backtest_source[
    [
        c for c in backtest_columns
        if c in backtest_source.columns
    ]
].copy()


# ============================================================
# 11. REMOVE DUPLICATE TRANSACTIONS
# ============================================================
#
# We should NOT silently aggregate duplicates.
#
# For production candidates, a transaction should appear once.
# If duplicates exist, STOP and investigate rather than inventing
# a deduplication rule.

production_duplicates = production_export[
    production_export["TRANSACTION_NUMBER"].duplicated(
        keep=False
    )
]

backtest_duplicates = backtest_export[
    backtest_export["TRANSACTION_NUMBER"].duplicated(
        keep=False
    )
]

print("\nDuplicate production transactions:",
      len(production_duplicates))

print("Duplicate backtest transactions:",
      len(backtest_duplicates))

assert len(production_duplicates) == 0, (
    "Duplicate TRANSACTION_NUMBER found in production export. "
    "Do not export until resolved."
)


# Backtest duplicates can legitimately exist depending on how the
# target transaction universe was constructed, so we report them
# rather than silently changing the validated backtest.

if len(backtest_duplicates) > 0:
    print(
        "\nWARNING: Backtest contains duplicate transaction numbers."
    )
    print(
        "This has NOT been deduplicated."
    )


# ============================================================
# 12. TYPE NORMALIZATION
# ============================================================

numeric_production_cols = [
    "TRANS_VALUE",
    "price_per_sqft",
    "fair_value",
    "valuation_gap_aed",
    "valuation_gap_pct",
    "risk_adjusted_gap_pct",
    "opportunity_score",
    "confidence_score",
    "current_ppsf",
    "future_30d_ppsf",
    "future_30d_comps",
]

for col in numeric_production_cols:
    if col in production_export.columns:
        production_export[col] = pd.to_numeric(
            production_export[col],
            errors="coerce"
        )


numeric_backtest_cols = [
    "current_ppsf",
    "future_30d_ppsf",
    "future_30d_comps",
    "forward_30d_return_pct",
    "opportunity_score",
    "risk_adjusted_gap_pct",
    "confidence_score",
]

for col in numeric_backtest_cols:
    if col in backtest_export.columns:
        backtest_export[col] = pd.to_numeric(
            backtest_export[col],
            errors="coerce"
        )


# ============================================================
# 13. FINAL SIGNAL COUNTS
# ============================================================

production_signal_counts = (
    production_export["final_signal"]
    .value_counts()
    .to_dict()
)

backtest_signal_counts = (
    backtest_export["final_signal"]
    .value_counts()
    .to_dict()
)


print("\n============================================================")
print("PRODUCTION SIGNAL COUNTS")
print("============================================================")

print(
    production_export["final_signal"]
    .value_counts()
)


print("\n============================================================")
print("BACKTEST SIGNAL COUNTS")
print("============================================================")

print(
    backtest_export["final_signal"]
    .value_counts()
)


# ============================================================
# 14. PRODUCTION SANITY CHECK
# ============================================================

production_conservative = (
    production_export["final_signal"]
    == "CONSERVATIVE OPPORTUNITY"
)

production_broad = (
    production_export["final_signal"]
    == "BROAD OPPORTUNITY"
)

print("\n============================================================")
print("PRODUCTION SANITY CHECK")
print("============================================================")

print(
    "Production candidates:",
    len(production_export)
)

print(
    "Conservative:",
    production_conservative.sum()
)

print(
    "Broad:",
    production_broad.sum()
)

print(
    "No signal:",
    (
        production_export["final_signal"]
        == "NO SIGNAL"
    ).sum()
)


# ============================================================
# 15. VERIFY PRODUCTION RULE
# ============================================================

expected_conservative = (
    production_export["opportunity"]
    .astype(str)
    .str.upper()
    .eq("HIGH OPPORTUNITY")
    &
    (
        production_export["confidence_score"]
        >= 80
    )
)

expected_broad = (
    production_export["opportunity"]
    .astype(str)
    .str.upper()
    .eq("HIGH OPPORTUNITY")
    &
    (
        production_export["confidence_score"]
        < 80
    )
)

actual_conservative = (
    production_export["final_signal"]
    == "CONSERVATIVE OPPORTUNITY"
)

actual_broad = (
    production_export["final_signal"]
    == "BROAD OPPORTUNITY"
)

assert (
    expected_conservative == actual_conservative
).all(), "Conservative signal rule mismatch."

assert (
    expected_broad == actual_broad
).all(), "Broad signal rule mismatch."


print("\nProduction rule verification: PASSED")


# ============================================================
# 16. BACKTEST SUMMARY FOR METADATA
# ============================================================

bt_conservative = backtest_export[
    backtest_export["final_signal"]
    == "CONSERVATIVE OPPORTUNITY"
].copy()

bt_broad = backtest_export[
    backtest_export["final_signal"]
    == "BROAD OPPORTUNITY"
].copy()

bt_non_signal = backtest_export[
    backtest_export["final_signal"]
    == "NO SIGNAL"
].copy()


def performance_summary(df):
    if len(df) == 0:
        return {
            "transactions": 0,
            "median_return_pct": None,
            "mean_return_pct": None,
            "positive_rate_pct": None,
            "above_10_pct_rate": None,
        }

    returns = pd.to_numeric(
        df["forward_30d_return_pct"],
        errors="coerce"
    ).dropna()

    return {
        "transactions": int(len(returns)),
        "median_return_pct": float(returns.median()),
        "mean_return_pct": float(returns.mean()),
        "positive_rate_pct": float(
            (returns > 0).mean() * 100
        ),
        "above_10_pct_rate": float(
            (returns > 10).mean() * 100
        ),
    }


backtest_summary = {
    "conservative": performance_summary(
        bt_conservative
    ),
    "broad": performance_summary(
        bt_broad
    ),
    "no_signal": performance_summary(
        bt_non_signal
    ),
}


# ============================================================
# 17. DATA DATE RANGES
# ============================================================

production_dates = pd.to_datetime(
    production_export.get(
        "target_date",
        pd.Series(dtype="datetime64[ns]")
    ),
    errors="coerce"
)

backtest_dates = pd.to_datetime(
    backtest_export["target_date"],
    errors="coerce"
)


def date_range_dict(series):
    series = series.dropna()

    if len(series) == 0:
        return {
            "start": None,
            "end": None,
        }

    return {
        "start": series.min().isoformat(),
        "end": series.max().isoformat(),
    }


# ============================================================
# 18. METADATA
# ============================================================

metadata = {
    "signal_version": "v1",

    "production_rule": {
        "conservative": {
            "opportunity": "HIGH OPPORTUNITY",
            "confidence_min": 80,
        },
        "broad": {
            "opportunity": "HIGH OPPORTUNITY",
            "confidence_max": 80,
        },
        "no_signal": "ALL OTHER TRANSACTIONS",
    },

    "production": {
        "rows": int(len(production_export)),
        "conservative": int(production_conservative.sum()),
        "broad": int(production_broad.sum()),
        "no_signal": int(
            (
                production_export["final_signal"]
                == "NO SIGNAL"
            ).sum()
        ),
        "date_range": date_range_dict(
            production_dates
        ),
    },

    "backtest": {
        "rows": int(len(backtest_export)),
        "date_range": date_range_dict(
            backtest_dates
        ),
        "signal_counts": {
            k: int(v)
            for k, v in backtest_signal_counts.items()
        },
        "performance": backtest_summary,
    },

    "model": {
        "type": "XGBoost",
        "objective": "reg:squarederror",
        "feature_count": 12,
        "learning_rate": 0.03,
        "max_depth": 6,
        "n_estimators": 2000,
        "min_child_weight": 10,
        "subsample": 0.8,
        "colsample_bytree": 0.8,
        "reg_alpha": 0.1,
        "reg_lambda": 5.0,
        "random_state": 42,
    },

    "dashboard": {
        "purpose": (
            "Production opportunity monitoring and "
            "validated signal analysis"
        ),
        "horizon_days": 30,
    },
}


# ============================================================
# 19. EXPORT CSV
# ============================================================

production_path = (
    OUTPUT_DIR /
    "production_candidates.csv"
)

backtest_path = (
    OUTPUT_DIR /
    "backtest_30d.csv"
)

metadata_path = (
    OUTPUT_DIR /
    "metadata.json"
)


production_export.to_csv(
    production_path,
    index=False
)

backtest_export.to_csv(
    backtest_path,
    index=False
)


# ============================================================
# 20. EXPORT METADATA JSON
# ============================================================

with open(
    metadata_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=2
    )


# ============================================================
# 21. VERIFY FILES
# ============================================================

print("\n============================================================")
print("EXPORT COMPLETE")
print("============================================================")

print("\nProduction dataset:")
print(production_path.resolve())
print("Shape:", production_export.shape)

print("\nBacktest dataset:")
print(backtest_path.resolve())
print("Shape:", backtest_export.shape)

print("\nMetadata:")
print(metadata_path.resolve())

print("\n------------------------------------------------------------")
print("Production signal counts")
print("------------------------------------------------------------")
print(
    production_export["final_signal"]
    .value_counts()
)

print("\n------------------------------------------------------------")
print("Backtest signal counts")
print("------------------------------------------------------------")
print(
    backtest_export["final_signal"]
    .value_counts()
)

print("\n------------------------------------------------------------")
print("Files exist")
print("------------------------------------------------------------")

print(
    "production_candidates.csv:",
    production_path.exists()
)

print(
    "backtest_30d.csv:",
    backtest_path.exists()
)

print(
    "metadata.json:",
    metadata_path.exists()
)

print("\n============================================================")
print("STREAMLIT DATA LAYER READY")
print("============================================================")
```
